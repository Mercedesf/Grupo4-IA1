````python
# ==============================================================================
# PROYECTO PONG CON CONTROLADORES DIFUSOS Y MENÚ DE SELECCIÓN
# Adaptado para Jupyter Notebook (Con CCD y Selección de Modos)
# ==============================================================================

import sys
import os
import math
import random
from enum import Enum, auto
from typing import Tuple, Dict, List, Any, Callable
import numpy as np

# Configuración de Pygame para Jupyter Notebook
os.environ['PYGAME_HIDE_SUPPORT_PROMPT'] = "hide"
import pygame

# ==============================================================================
# 1. MÓDULO DE CONFIGURACIÓN GLOBAL
# ==============================================================================
ANCHO_PANTALLA: int = 1100  # Incluye panel lateral de telemetría
ANCHO_CANCHA: int = 800
ALTO_CANCHA: int = 600

# Palas
ANCHO_PALA: int = 15
ALTO_PALA: int = 90
VELOCIDAD_PALA_BASE: float = 520.0  # Píxeles por segundo

# Pelota
RADIO_PELOTA: int = 9
VELOCIDAD_INICIAL_PELOTA_X: float = 380.0
VELOCIDAD_INICIAL_PELOTA_Y: float = 220.0
VELOCIDAD_MIN_PELOTA: float = 300.0
VELOCIDAD_MAX_PELOTA: float = 1400.0
FACTOR_INCREMENTO_VELOCIDAD: float = 1.06  

# Puntuación
PUNTAJE_MAXIMO: int = 5

# Colores (RGB)
COLOR_FONDO: Tuple[int, int, int] = (18, 18, 24)
COLOR_CANCHA: Tuple[int, int, int] = (30, 30, 42)
COLOR_TEXTO: Tuple[int, int, int] = (240, 240, 240)
COLOR_TEXTO_MUTED: Tuple[int, int, int] = (160, 160, 180)
COLOR_PALA_IZQ: Tuple[int, int, int] = (50, 150, 255)
COLOR_PALA_DER: Tuple[int, int, int] = (255, 90, 90)
COLOR_PELOTA: Tuple[int, int, int] = (255, 220, 0)
COLOR_PANEL_DEBUG: Tuple[int, int, int] = (10, 10, 15)

FPS: int = 60

# ==============================================================================
# 2. ENTIDADES Y FÍSICA AVANZADA (CCD SIN TUNNELING)
# ==============================================================================
class Ball:
    """Representa la pelota con registro de posición previa para Sweep-Test/CCD."""
    def __init__(self) -> None:
        self.radius: int = RADIO_PELOTA
        self.x: float = ANCHO_CANCHA / 2.0
        self.y: float = ALTO_CANCHA / 2.0
        self.prev_x: float = self.x
        self.prev_y: float = self.y
        self.vx: float = VELOCIDAD_INICIAL_PELOTA_X
        self.vy: float = VELOCIDAD_INICIAL_PELOTA_Y
        self.reset(dir_x=1)

    def reset(self, dir_x: int = 1) -> None:
        self.x = ANCHO_CANCHA / 2.0
        self.y = ALTO_CANCHA / 2.0
        self.prev_x = self.x
        self.prev_y = self.y
        angulo_rad = random.uniform(-math.pi / 6, math.pi / 6)
        rapidez_base = math.hypot(VELOCIDAD_INICIAL_PELOTA_X, VELOCIDAD_INICIAL_PELOTA_Y)
        self.vx = dir_x * rapidez_base * math.cos(angulo_rad)
        self.vy = rapidez_base * math.sin(angulo_rad)

    def update(self, dt: float) -> None:
        self.prev_x = self.x
        self.prev_y = self.y
        self.x += self.vx * dt
        self.y += self.vy * dt

    def boost_speed(self, factor: float) -> None:
        rapidez_actual = math.hypot(self.vx, self.vy)
        nueva_rapidez = min(rapidez_actual * factor, VELOCIDAD_MAX_PELOTA)
        nueva_rapidez = max(nueva_rapidez, VELOCIDAD_MIN_PELOTA)
        if rapidez_actual > 0:
            escala = nueva_rapidez / rapidez_actual
            self.vx *= escala
            self.vy *= escala

class Paddle:
    """Representa las palas desplazables únicamente en vertical."""
    def __init__(self, x: float, y: float) -> None:
        self.x: float = x
        self.y: float = y
        self.width: int = ANCHO_PALA
        self.height: int = ALTO_PALA
        self.speed: float = VELOCIDAD_PALA_BASE
        self.vy_cmd: float = 0.0

    def move(self, direction: float, dt: float) -> None:
        self.vy_cmd = max(-1.0, min(1.0, direction))
        self.y += self.vy_cmd * self.speed * dt
        mitad_altura = self.height / 2.0
        if self.y - mitad_altura < 0:
            self.y = mitad_altura
        elif self.y + mitad_altura > ALTO_CANCHA:
            self.y = ALTO_CANCHA - mitad_altura

    @property
    def center_y(self) -> float:
        return self.y

    @property
    def rect_bounds(self) -> Tuple[float, float, float, float]:
        return (
            self.x - self.width / 2.0,
            self.y - self.height / 2.0,
            self.x + self.width / 2.0,
            self.y + self.height / 2.0
        )

# Detección Físicamente Precisa
def check_wall_collision(ball: Ball) -> bool:
    if ball.y - ball.radius <= 0:
        ball.y = ball.radius + 1.0
        ball.vy = abs(ball.vy)
        ball.boost_speed(FACTOR_INCREMENTO_VELOCIDAD)
        return True
    elif ball.y + ball.radius >= ALTO_CANCHA:
        ball.y = ALTO_CANCHA - ball.radius - 1.0
        ball.vy = -abs(ball.vy)
        ball.boost_speed(FACTOR_INCREMENTO_VELOCIDAD)
        return True
    return False

def check_paddle_collision_ccd(ball: Ball, paddle: Paddle, is_left_paddle: bool) -> bool:
    left, top, right, bottom = paddle.rect_bounds
    
    # 1. Colisión AABB Estándar
    closest_x = max(left, min(ball.x, right))
    closest_y = max(top, min(ball.y, bottom))
    if ((ball.x - closest_x)**2 + (ball.y - closest_y)**2) <= ball.radius**2:
        _resolve_paddle_impact(ball, paddle, is_left_paddle)
        return True

    # 2. Continuous Collision Detection por Trazo Lineal (CCD)
    plane_x = right + ball.radius if is_left_paddle else left - ball.radius
    crossed_plane = (ball.prev_x >= plane_x >= ball.x) if is_left_paddle else (ball.prev_x <= plane_x <= ball.x)
    
    if crossed_plane and (ball.x != ball.prev_x):
        t = (plane_x - ball.prev_x) / (ball.x - ball.prev_x)
        y_at_crossing = ball.prev_y + t * (ball.y - ball.prev_y)
        if (top - ball.radius) <= y_at_crossing <= (bottom + ball.radius):
            _resolve_paddle_impact(ball, paddle, is_left_paddle)
            return True

    return False

def _resolve_paddle_impact(ball: Ball, paddle: Paddle, is_left_paddle: bool) -> None:
    left, top, right, bottom = paddle.rect_bounds
    if is_left_paddle:
        ball.x = right + ball.radius + 1.0
        ball.vx = abs(ball.vx)
    else:
        ball.x = left - ball.radius - 1.0
        ball.vx = -abs(ball.vx)
        
    impacto_relativo = (ball.y - paddle.y) / (paddle.height / 2.0)
    ball.vy += impacto_relativo * 140.0
    ball.boost_speed(FACTOR_INCREMENTO_VELOCIDAD)

def check_goal(ball: Ball) -> Tuple[bool, int]:
    if ball.x + ball.radius < 0:
        return True, 2  # Punto para la derecha
    elif ball.x - ball.radius > ANCHO_CANCHA:
        return True, 1  # Punto para la izquierda
    return False, 0

class GameMode(Enum):
    PLAYER_VS_PC = auto()
    PLAYER_VS_PLAYER = auto()
    PC_VS_PC = auto()

class GameState(Enum):
    MENU = auto()
    PLAYING = auto()
    GAME_OVER = auto()

class GameEngine:
    """Motor de estado del juego actualizado con soporte para 3 modos de juego."""
    def __init__(self) -> None:
        self.state: GameState = GameState.MENU
        self.mode: GameMode = GameMode.PLAYER_VS_PC
        self.ball: Ball = Ball()
        self.paddle_left: Paddle = Paddle(x=30, y=ALTO_CANCHA / 2.0)
        self.paddle_right: Paddle = Paddle(x=ANCHO_CANCHA - 30, y=ALTO_CANCHA / 2.0)
        self.score_left: int = 0
        self.score_right: int = 0
        self.winner_text: str = ""

    def start_game(self, mode: GameMode) -> None:
        """Inicia una nueva partida con el modo de juego seleccionado."""
        self.mode = mode
        self.restart()

    def update(self, dt: float, cmd_left: float, cmd_right: float) -> None:
        if self.state != GameState.PLAYING:
            return

        self.paddle_left.move(cmd_left, dt)
        self.paddle_right.move(cmd_right, dt)

        # Sub-stepping para física a alta velocidad
        speed = math.hypot(self.ball.vx, self.ball.vy)
        sub_steps = max(1, int(speed / 300.0))
        sub_dt = dt / sub_steps

        for _ in range(sub_steps):
            self.ball.update(sub_dt)
            check_wall_collision(self.ball)
            c_left = check_paddle_collision_ccd(self.ball, self.paddle_left, is_left_paddle=True)
            c_right = check_paddle_collision_ccd(self.ball, self.paddle_right, is_left_paddle=False)
            if c_left or c_right:
                break

        is_goal, winner = check_goal(self.ball)
        if is_goal:
            if winner == 1:
                self.score_left += 1
                self.ball.reset(dir_x=-1)
            else:
                self.score_right += 1
                self.ball.reset(dir_x=1)
            self._check_game_over()

    def _check_game_over(self) -> None:
        if self.score_left >= PUNTAJE_MAXIMO:
            self.state = GameState.GAME_OVER
            if self.mode == GameMode.PLAYER_VS_PC:
                self.winner_text = "¡JUGADOR 1 GANA!"
            elif self.mode == GameMode.PLAYER_VS_PLAYER:
                self.winner_text = "¡JUGADOR 1 GANA!"
            else:
                self.winner_text = "¡AGENTE MAMDANI GANA!"
        elif self.score_right >= PUNTAJE_MAXIMO:
            self.state = GameState.GAME_OVER
            if self.mode == GameMode.PLAYER_VS_PC:
                self.winner_text = "¡AGENTE DIFUSO GANA!"
            elif self.mode == GameMode.PLAYER_VS_PLAYER:
                self.winner_text = "¡JUGADOR 2 GANA!"
            else:
                self.winner_text = "¡AGENTE TAKAGI-SUGENO GANA!"

    def restart(self) -> None:
        self.score_left = 0
        self.score_right = 0
        self.state = GameState.PLAYING
        self.winner_text = ""
        self.ball.reset()
        self.paddle_left.y = ALTO_CANCHA / 2.0
        self.paddle_right.y = ALTO_CANCHA / 2.0

# ==============================================================================
# 3. CONTROLADORES DIFUSOS
# ==============================================================================
def triangular(a: float, b: float, c: float) -> Callable[[float], float]:
    def mf(x: float) -> float:
        if x <= a or x >= c:
            return 0.0
        if x == b:
            return 1.0
        if a < x < b:
            return (x - a) / (b - a)
        return (c - x) / (c - b)
    return mf

def trapezoidal(a: float, b: float, c: float, d: float) -> Callable[[float], float]:
    def mf(x: float) -> float:
        if x <= a or x >= d:
            return 0.0
        if b <= x <= c:
            return 1.0
        if a < x < b:
            return (x - a) / (b - a)
        return (d - x) / (d - c)
    return mf

BASE_REGLAS: List[Tuple[str, str, str]] = [
    ("MUY_ABAJO", "SUBIENDO_RAPIDO", "RAPIDO_ARRIBA"),
    ("MUY_ABAJO", "NEUTRO",          "RAPIDO_ARRIBA"),
    ("MUY_ABAJO", "BAJANDO_RAPIDO",  "ARRIBA"),
    ("ABAJO",     "SUBIENDO_RAPIDO", "RAPIDO_ARRIBA"),
    ("ABAJO",     "NEUTRO",          "ARRIBA"),
    ("ABAJO",     "BAJANDO_RAPIDO",  "QUIETO"),
    ("CENTRO",    "SUBIENDO_RAPIDO", "ARRIBA"),
    ("CENTRO",    "NEUTRO",          "QUIETO"),
    ("CENTRO",    "BAJANDO_RAPIDO",  "ABAJO"),
    ("ARRIBA",    "SUBIENDO_RAPIDO", "QUIETO"),
    ("ARRIBA",    "NEUTRO",          "ABAJO"),
    ("ARRIBA",    "BAJANDO_RAPIDO",  "RAPIDO_ABAJO"),
    ("MUY_ARRIBA", "SUBIENDO_RAPIDO", "ABAJO"),
    ("MUY_ARRIBA", "NEUTRO",          "RAPIDO_ABAJO"),
    ("MUY_ARRIBA", "BAJANDO_RAPIDO",  "RAPIDO_ABAJO"),
]

class MamdaniInferenceEngine:
    def __init__(self, step_resolution: int = 80) -> None:
        self.resolution = step_resolution

    def evaluate(
        self,
        inputs: Dict[str, float],
        input_mfs: Dict[str, Dict[str, Callable[[float], float]]],
        output_mfs: Dict[str, Callable[[float], float]],
        rules: List[Tuple[str, str, str]],
        output_range: Tuple[float, float]
    ) -> Tuple[float, Dict[str, float], List[float]]:
        fuzzified = {var: {s: mf(val) for s, mf in mfs.items()} for var, val in inputs.items() for mfs in [input_mfs[var]]}
        rule_firings: List[float] = []
        rule_outputs: List[Tuple[float, Callable[[float], float]]] = []

        for e_set, v_set, out_set in rules:
            w = min(fuzzified["error"][e_set], fuzzified["vel_y"][v_set])
            rule_firings.append(w)
            rule_outputs.append((w, output_mfs[out_set]))

        y_samples = np.linspace(output_range[0], output_range[1], self.resolution)
        aggregated_mf = np.zeros_like(y_samples)

        for i, y in enumerate(y_samples):
            mu_max = 0.0
            for w, mf_out in rule_outputs:
                mu_clipped = min(w, mf_out(y))
                if mu_clipped > mu_max:
                    mu_max = mu_clipped
            aggregated_mf[i] = mu_max

        sum_mu = np.sum(aggregated_mf)
        defuzzified = 0.0 if sum_mu == 0 else float(np.sum(y_samples * aggregated_mf) / sum_mu)
        fuzz_flat = {f"{k}_{k2}": v2 for k, v in fuzzified.items() for k2, v2 in v.items()}
        return defuzzified, fuzz_flat, rule_firings

class SugenoInferenceEngine:
    def evaluate(
        self,
        inputs: Dict[str, float],
        input_mfs: Dict[str, Dict[str, Callable[[float], float]]],
        output_values: Dict[str, float],
        rules: List[Tuple[str, str, str]]
    ) -> float:
        fuzzified = {var: {s: mf(val) for s, mf in mfs.items()} for var, val in inputs.items() for mfs in [input_mfs[var]]}
        numerator, denominator = 0.0, 0.0

        for e_set, v_set, out_set_key in rules:
            w = min(fuzzified["error"][e_set], fuzzified["vel_y"][v_set])
            z = output_values[out_set_key]
            numerator += w * z
            denominator += w

        return 0.0 if denominator == 0 else numerator / denominator

class MamdaniAgent:
    def __init__(self) -> None:
        self.engine = MamdaniInferenceEngine()
        self.input_mfs = {
            "error": {
                "MUY_ABAJO": trapezoidal(-700, -600, -300, -100),
                "ABAJO":     triangular(-200, -80, 0),
                "CENTRO":    triangular(-35, 0, 35),
                "ARRIBA":    triangular(0, 80, 200),
                "MUY_ARRIBA": trapezoidal(100, 300, 600, 700)
            },
            "vel_y": {
                "SUBIENDO_RAPIDO": trapezoidal(-1400, -1000, -500, -100),
                "NEUTRO":          triangular(-200, 0, 200),
                "BAJANDO_RAPIDO":  trapezoidal(100, 500, 1000, 1400)
            }
        }
        self.output_mfs = {
            "RAPIDO_ARRIBA": triangular(-1.0, -1.0, -0.5),
            "ARRIBA":        triangular(-0.7, -0.35, 0.0),
            "QUIETO":        triangular(-0.15, 0.0, 0.15),
            "ABAJO":         triangular(0.0, 0.35, 0.7),
            "RAPIDO_ABAJO":  triangular(0.5, 1.0, 1.0)
        }
        self.last_debug: Dict[str, Any] = {}

    def compute(self, ball_y: float, paddle_y: float, ball_vy: float) -> float:
        error_vert = ball_y - paddle_y
        out_cmd, fuzzified, firings = self.engine.evaluate(
            inputs={"error": error_vert, "vel_y": ball_vy},
            input_mfs=self.input_mfs,
            output_mfs=self.output_mfs,
            rules=BASE_REGLAS,
            output_range=(-1.0, 1.0)
        )
        self.last_debug = {
            "error": error_vert, "ball_vy": ball_vy, "output": out_cmd,
            "fuzzified": fuzzified, "firings": firings
        }
        return out_cmd

    def get_debug_info(self) -> Dict[str, Any]:
        return self.last_debug

class SugenoAgent:
    def __init__(self) -> None:
        self.engine = SugenoInferenceEngine()
        self.input_mfs = {
            "error": {
                "MUY_ABAJO": trapezoidal(-700, -600, -250, -80),
                "ABAJO":     triangular(-150, -60, 0),
                "CENTRO":    triangular(-20, 0, 20),
                "ARRIBA":    triangular(0, 60, 150),
                "MUY_ARRIBA": trapezoidal(80, 250, 600, 700)
            },
            "vel_y": {
                "SUBIENDO_RAPIDO": trapezoidal(-1400, -1000, -400, -80),
                "NEUTRO":          triangular(-150, 0, 150),
                "BAJANDO_RAPIDO":  trapezoidal(80, 400, 1000, 1400)
            }
        }
        self.output_values = {
            "RAPIDO_ARRIBA": -1.0, "ARRIBA": -0.5, "QUIETO": 0.0,
            "ABAJO": 0.5, "RAPIDO_ABAJO": 1.0
        }

    def compute(self, ball_y: float, paddle_y: float, ball_vy: float) -> float:
        return self.engine.evaluate(
            inputs={"error": ball_y - paddle_y, "vel_y": ball_vy},
            input_mfs=self.input_mfs,
            output_values=self.output_values,
            rules=BASE_REGLAS
        )

# ==============================================================================
# 4. RENDERIZADO E INTERFAZ GRÁFICA
# ==============================================================================
class Renderer:
    def __init__(self, surface: pygame.Surface) -> None:
        self.surface: pygame.Surface = surface
        self.font_title = pygame.font.SysFont("Consolas", 38, bold=True)
        self.font_score = pygame.font.SysFont("Consolas", 48, bold=True)
        self.font_ui = pygame.font.SysFont("Consolas", 16)
        self.font_debug = pygame.font.SysFont("Consolas", 13)

    def render(self, game: GameEngine, debug_info: Dict[str, Any]) -> None:
        self.surface.fill(COLOR_FONDO)

        if game.state == GameState.MENU:
            self._render_menu()
            return

        # Dibujar Cancha y juego en ejecución
        pygame.draw.rect(self.surface, COLOR_CANCHA, (0, 0, ANCHO_CANCHA, ALTO_CANCHA))
        for y in range(0, ALTO_CANCHA, 20):
            pygame.draw.rect(self.surface, (70, 70, 90), (ANCHO_CANCHA // 2 - 2, y, 4, 10))

        # Palas y Pelota
        left_r = game.paddle_left.rect_bounds
        right_r = game.paddle_right.rect_bounds
        pygame.draw.rect(self.surface, COLOR_PALA_IZQ, (left_r[0], left_r[1], game.paddle_left.width, game.paddle_left.height), border_radius=4)
        pygame.draw.rect(self.surface, COLOR_PALA_DER, (right_r[0], right_r[1], game.paddle_right.width, game.paddle_right.height), border_radius=4)
        pygame.draw.circle(self.surface, COLOR_PELOTA, (int(game.ball.x), int(game.ball.y)), game.ball.radius)

        # Marcador
        txt_score = self.font_score.render(f"{game.score_left}   {game.score_right}", True, COLOR_TEXTO)
        self.surface.blit(txt_score, (ANCHO_CANCHA // 2 - txt_score.get_width() // 2, 20))

        # Telemetría Lateral
        self._render_debug_panel(debug_info, game.mode)

        if game.state == GameState.GAME_OVER:
            self._render_game_over(game.winner_text)

    def _render_menu(self) -> None:
        """Renderiza el menú de selección inicial."""
        # Título
        title = self.font_title.render("PONG FUZZY CONTROL", True, (0, 255, 200))
        subtitle = self.font_ui.render("Selecciona el modo de juego para comenzar:", True, COLOR_TEXTO)
        
        self.surface.blit(title, (ANCHO_PANTALLA // 2 - title.get_width() // 2, 80))
        self.surface.blit(subtitle, (ANCHO_PANTALLA // 2 - subtitle.get_width() // 2, 140))

        # Opciones
        options = [
            ("[ 1 ] JUGADOR vs PC", "Pala Izquierda: Teclas W / S  |  Pala Derecha: Agente Difuso Mamdani"),
            ("[ 2 ] JUGADOR vs JUGADOR", "Jugador 1: Teclas W / S  |  Jugador 2: Flechas Arriba / Abajo"),
            ("[ 3 ] PC vs PC", "Agente Izquierdo: Mamdani  |  Agente Derecho: Takagi-Sugeno")
        ]

        start_y = 220
        for idx, (op_title, op_desc) in enumerate(options):
            box_rect = pygame.Rect(ANCHO_PANTALLA // 2 - 320, start_y + idx * 100, 640, 80)
            pygame.draw.rect(self.surface, (30, 35, 50), box_rect, border_radius=8)
            pygame.draw.rect(self.surface, (60, 120, 200), box_rect, width=2, border_radius=8)

            txt_t = self.font_ui.render(op_title, True, (255, 220, 0))
            txt_d = self.font_debug.render(op_desc, True, COLOR_TEXTO_MUTED)

            self.surface.blit(txt_t, (box_rect.x + 20, box_rect.y + 18))
            self.surface.blit(txt_d, (box_rect.x + 20, box_rect.y + 48))

        footer = self.font_debug.render("Presiona la tecla 1, 2 o 3 para iniciar", True, COLOR_TEXTO)
        self.surface.blit(footer, (ANCHO_PANTALLA // 2 - footer.get_width() // 2, 540))

    def _render_debug_panel(self, debug_info: Dict[str, Any], mode: GameMode) -> None:
        panel_x = ANCHO_CANCHA
        pygame.draw.rect(self.surface, COLOR_PANEL_DEBUG, (panel_x, 0, ANCHO_PANTALLA - ANCHO_CANCHA, ALTO_CANCHA))
        pygame.draw.line(self.surface, (60, 60, 80), (panel_x, 0), (panel_x, ALTO_CANCHA), 2)

        y_offset = 20
        title = self.font_ui.render("TELEMETRÍA DIFUSA", True, (0, 255, 200))
        self.surface.blit(title, (panel_x + 15, y_offset))
        y_offset += 35

        lines = [
            f"Modo: {mode.name}",
            f"Error: {debug_info.get('error', 0.0):.1f} px",
            f"Vel. Y Pelota: {debug_info.get('ball_vy', 0.0):.1f} px/s",
            f"Salida Cmd: {debug_info.get('output', 0.0):.3f}",
            "-" * 28,
            "Conjuntos Difusos Activos:"
        ]

        for line in lines:
            t = self.font_debug.render(line, True, COLOR_TEXTO)
            self.surface.blit(t, (panel_x + 15, y_offset))
            y_offset += 20

        fuzz = debug_info.get("fuzzified", {})
        for name, val in fuzz.items():
            if val > 0.01:
                lbl = self.font_debug.render(f"  {name}: {val:.2f}", True, (255, 200, 100))
                self.surface.blit(lbl, (panel_x + 15, y_offset))
                y_offset += 18

    def _render_game_over(self, winner_text: str) -> None:
        s = pygame.Surface((ANCHO_CANCHA, ALTO_CANCHA), pygame.SRCALPHA)
        s.fill((0, 0, 0, 200))
        self.surface.blit(s, (0, 0))

        txt1 = self.font_score.render("FIN DE LA PARTIDA", True, (255, 215, 0))
        txt2 = self.font_ui.render(winner_text, True, COLOR_TEXTO)
        txt3 = self.font_ui.render("Presione 'R' para Reiniciar | 'M' para Volver al Menú", True, (180, 180, 180))

        self.surface.blit(txt1, (ANCHO_CANCHA // 2 - txt1.get_width() // 2, 200))
        self.surface.blit(txt2, (ANCHO_CANCHA // 2 - txt2.get_width() // 2, 280))
        self.surface.blit(txt3, (ANCHO_CANCHA // 2 - txt3.get_width() // 2, 350))

# ==============================================================================
# 5. BUCLE PRINCIPAL DE JUPYTER
# ==============================================================================
def run_game_in_jupyter():
    pygame.init()
    pygame.font.init()
    
    screen = pygame.display.set_mode((ANCHO_PANTALLA, ALTO_CANCHA))
    pygame.display.set_caption("Pong Fuzzy Control - Menú de Selección")
    clock = pygame.time.Clock()

    game = GameEngine()
    renderer = Renderer(screen)
    agent_mamdani = MamdaniAgent()
    agent_sugeno = SugenoAgent()

    running = True
    try:
        while running:
            dt = clock.tick(FPS) / 1000.0
            if dt > 0.05:
                dt = 0.05

            # Control de Eventos
            for event in pygame.event.get():
                if event.type == pygame.QUIT:
                    running = False
                elif event.type == pygame.KEYDOWN:
                    # Controles en estado MENÚ
                    if game.state == GameState.MENU:
                        if event.key == pygame.K_1 or event.key == pygame.K_KP1:
                            game.start_game(GameMode.PLAYER_VS_PC)
                        elif event.key == pygame.K_2 or event.key == pygame.K_KP2:
                            game.start_game(GameMode.PLAYER_VS_PLAYER)
                        elif event.key == pygame.K_3 or event.key == pygame.K_KP3:
                            game.start_game(GameMode.PC_VS_PC)
                    # Controles globales en juego
                    else:
                        if event.key == pygame.K_r:
                            game.restart()
                        elif event.key == pygame.K_m:
                            game.state = GameState.MENU

            # Cálculo de Comandos de Control según el Modo Activo
            cmd_left, cmd_right = 0.0, 0.0
            if game.state == GameState.PLAYING:
                keys = pygame.key.get_pressed()

                # Modo 1: Jugador vs PC
                if game.mode == GameMode.PLAYER_VS_PC:
                    if keys[pygame.K_w]:
                        cmd_left = -1.0
                    elif keys[pygame.K_s]:
                        cmd_left = 1.0
                    
                    cmd_right = agent_mamdani.compute(game.ball.y, game.paddle_right.center_y, game.ball.vy)

                # Modo 2: Jugador vs Jugador
                elif game.mode == GameMode.PLAYER_VS_PLAYER:
                    # Jugador 1 (W / S)
                    if keys[pygame.K_w]:
                        cmd_left = -1.0
                    elif keys[pygame.K_s]:
                        cmd_left = 1.0

                    # Jugador 2 (Flecha Arriba / Flecha Abajo)
                    if keys[pygame.K_UP]:
                        cmd_right = -1.0
                    elif keys[pygame.K_DOWN]:
                        cmd_right = 1.0

                # Modo 3: PC vs PC
                elif game.mode == GameMode.PC_VS_PC:
                    cmd_left = agent_mamdani.compute(game.ball.y, game.paddle_left.center_y, game.ball.vy)
                    cmd_right = agent_sugeno.compute(game.ball.y, game.paddle_right.center_y, game.ball.vy)

            game.update(dt, cmd_left, cmd_right)
            renderer.render(game, agent_mamdani.get_debug_info())
            pygame.display.flip()

    finally:
        pygame.quit()

# Iniciar la aplicación en Jupyter Notebook
run_game_in_jupyter()
````
