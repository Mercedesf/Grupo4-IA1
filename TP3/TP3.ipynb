4. Encuentre el máximo de la función $f(x) = \frac{\sin(x)}{x + 0.1}$ en $x \in [-10; -6]$ con un error menor a $0.1$ utilizando el algoritmo _hill climbing_.

```python
import math
import random

# 1. Definimos la función matemática
def f(x):
    return math.sin(x) / (x + 0.1)

def hill_climbing(bounds, step_size):
    # 2. Elegimos un punto de inicio aleatorio dentro del dominio [-10, -6]
    current_x = random.uniform(bounds[0], bounds[1])
    current_y = f(current_x)
    
    print(f"Punto de inicio: x = {current_x:.4f}, f(x) = {current_y:.4f}")
    
    while True:
        # 3. Generamos los vecinos (izquierda y derecha) según el tamaño del paso
        left_x = current_x - step_size
        right_x = current_x + step_size
        
        # Filtramos para asegurarnos de no salirnos de los límites del dominio
        neighbors = []
        if left_x >= bounds[0]:
            neighbors.append(left_x)
        if right_x <= bounds[1]:
            neighbors.append(right_x)
            
        # 4. Evaluamos cuál es el mejor vecino
        best_neighbor_x = current_x
        best_neighbor_y = current_y
        
        for neighbor_x in neighbors:
            neighbor_y = f(neighbor_x)
            # Si el vecino es mejor que nuestro máximo actual, lo actualizamos
            if neighbor_y > best_neighbor_y:
                best_neighbor_y = neighbor_y
                best_neighbor_x = neighbor_x
                
        # 5. Condición de parada: Si ningún vecino mejora el valor actual, encontramos un máximo local
        if best_neighbor_x == current_x:
            break
            
        # Nos movemos al mejor vecino para la siguiente iteración
        current_x = best_neighbor_x
        current_y = best_neighbor_y
        
    return current_x, current_y

# --- Parámetros del problema ---
dominio = [-10, -6]
error_tolerado = 0.1
# Usamos un paso menor al error solicitado para garantizar precisión
paso = 0.01 

# Ejecutamos el algoritmo
max_x, max_y = hill_climbing(dominio, paso)

print("-" * 30)
print(f"Máximo encontrado en: x = {max_x:.4f}")
print(f"Valor máximo f(x) = {max_y:.4f}")
```

5. Diseñe e implemente un algoritmo de Recocido Simulado para que juegue contra usted al Ta-te-ti. Varíe los valores de temperatura inicial entre partidas, ¿qué diferencia observa cuando la temperatura es más alta con respecto a cuando la temperatura es más baja?

-Con T incial alta:Juega de manera estrategica anulando las combinaciones enemigas que intenten ganar.Y a la vez generando combinaciones propias que le hagan ganar

-Con T inicial muy baja:Busca rapidamente la casilla que le de mayor puntaje sin tener en cuenta la del enemigo por lo que se le puede ganar facilmente

```python
import tkinter as tk
from tkinter import messagebox
import random
import math

# =================================================================
# ⚙️ PARÁMETROS DEL ALGORITMO (¡MODIFICA ESTO!)
# =================================================================
# TEMPERATURA_INICIAL controla qué tan "loca" o "creativa" es la IA al principio.
# - Un valor ALTO (ej. 100.0 o 500.0): La IA explorará más y tomará decisiones 
#   al azar, incluso si parecen malas en ese momento.
# - Un valor BAJO (ej. 1.0 o 0.1): La IA será muy conservadora y casi siempre
#   elegirá el movimiento que le dé más puntos inmediatamente (como un algoritmo voraz).
TEMPERATURA_INICIAL = 10000000000  
# =================================================================

class TatetiGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("Tateti vs IA (Recocido Simulado)")
        
        self.jugador_humano = 'X'
        self.jugador_ia = 'O'
        self.tablero = [' '] * 9
        self.botones = []
        
        # Etiqueta de estado
        self.label_estado = tk.Label(root, text="¡Tu turno! Juegas con la X", font=('Arial', 14))
        self.label_estado.grid(row=0, column=0, columnspan=3, pady=10)
        
        # Crear la cuadrícula de botones 3x3
        for i in range(9):
            boton = tk.Button(root, text=' ', font=('Arial', 40, 'bold'), height=2, width=5,
                              command=lambda idx=i: self.click_humano(idx))
            boton.grid(row=(i//3)+1, column=i%3, padx=5, pady=5)
            self.botones.append(boton)
            
        # Botón para reiniciar
        self.boton_reinicio = tk.Button(root, text="Reiniciar Juego", font=('Arial', 12), command=self.reiniciar_juego)
        self.boton_reinicio.grid(row=4, column=0, columnspan=3, pady=10)

    def click_humano(self, idx):
        # Si la casilla está vacía
        if self.tablero[idx] == ' ':
            self.tablero[idx] = self.jugador_humano
            self.actualizar_botones()
            
            if self.verificar_victoria(self.tablero, self.jugador_humano):
                self.terminar_juego("¡Felicidades, ganaste!")
                return
            elif ' ' not in self.tablero:
                self.terminar_juego("¡Es un empate!")
                return
                
            # Turno de la IA
            self.label_estado.config(text="La IA está pensando...")
            self.root.update()
            
            # Un pequeño retraso para que parezca que "piensa"
            self.root.after(500, self.turno_ia)

    def turno_ia(self):
        movimiento = self.recocido_simulado(self.tablero, self.jugador_ia, self.jugador_humano)
        
        if movimiento is not None:
            self.tablero[movimiento] = self.jugador_ia
            self.actualizar_botones()
            
            if self.verificar_victoria(self.tablero, self.jugador_ia):
                self.terminar_juego("La IA te ha ganado. ¡Suerte a la próxima!")
                return
            elif ' ' not in self.tablero:
                self.terminar_juego("¡Es un empate!")
                return
                
            self.label_estado.config(text="¡Tu turno! Juegas con la X")

    def actualizar_botones(self):
        for i in range(9):
            self.botones[i].config(text=self.tablero[i])
            if self.tablero[i] != ' ':
                self.botones[i].config(state=tk.DISABLED, disabledforeground="black")

    def terminar_juego(self, mensaje):
        self.label_estado.config(text=mensaje)
        for boton in self.botones:
            boton.config(state=tk.DISABLED)
        messagebox.showinfo("Fin del Juego", mensaje)

    def reiniciar_juego(self):
        self.tablero = [' '] * 9
        for boton in self.botones:
            boton.config(text=' ', state=tk.NORMAL)
        self.label_estado.config(text="¡Tu turno! Juegas con la X")

    # --- LÓGICA DEL RECOCIDO SIMULADO ---
    
    def verificar_victoria(self, tablero, jugador):
        lineas = [
            [0, 1, 2], [3, 4, 5], [6, 7, 8], # Filas
            [0, 3, 6], [1, 4, 7], [2, 5, 8], # Columnas
            [0, 4, 8], [2, 4, 6]             # Diagonales
        ]
        for linea in lineas:
            if tablero[linea[0]] == tablero[linea[1]] == tablero[linea[2]] == jugador:
                return True
        return False

    def evaluar_movimiento(self, tablero, movimiento, jugador_ia, jugador_humano):
        tablero_temp = tablero.copy()
        
        # 1. ¿Gana la IA con este movimiento? (Máxima prioridad)
        tablero_temp[movimiento] = jugador_ia
        if self.verificar_victoria(tablero_temp, jugador_ia):
            return 1000
            
        # 2. ¿Bloquea al humano de ganar en el siguiente turno?
        tablero_temp[movimiento] = jugador_humano
        if self.verificar_victoria(tablero_temp, jugador_humano):
            return 500
            
        # 3. Estrategia posicional
        if movimiento == 4: # El centro es la mejor casilla
            return 50
        if movimiento in [0, 2, 6, 8]: # Las esquinas son la segunda mejor opción
            return 10
            
        return 0 # Bordes

    def recocido_simulado(self, tablero, jugador_ia, jugador_humano):
        movimientos_disponibles = [i for i, casilla in enumerate(tablero) if casilla == ' ']
        if not movimientos_disponibles:
            return None

        # Variables del algoritmo
        temperatura = TEMPERATURA_INICIAL # Tomamos la constante que definiste arriba
        temperatura_minima = 0.1
        tasa_enfriamiento = 0.9

        # 1. Empezamos con un movimiento aleatorio
        movimiento_actual = random.choice(movimientos_disponibles)
        puntaje_actual = self.evaluar_movimiento(tablero, movimiento_actual, jugador_ia, jugador_humano)

        # 2. Bucle de enfriamiento
        while temperatura > temperatura_minima:
            nuevo_movimiento = random.choice(movimientos_disponibles)
            nuevo_puntaje = self.evaluar_movimiento(tablero, nuevo_movimiento, jugador_ia, jugador_humano)

            # Diferencia de "energía"
            delta_e = nuevo_puntaje - puntaje_actual

            if delta_e > 0:
                # Es un movimiento mejor, lo aceptamos
                movimiento_actual = nuevo_movimiento
                puntaje_actual = nuevo_puntaje
            else:
                # Es un movimiento peor, lo aceptamos con cierta probabilidad (depende de la Temperatura)
                try:
                    probabilidad = math.exp(delta_e / temperatura)
                except OverflowError:
                    probabilidad = 0.0 # Prevención de error por números muy grandes
                    
                if random.random() < probabilidad:
                    movimiento_actual = nuevo_movimiento
                    puntaje_actual = nuevo_puntaje
            
            # Enfriamos el sistema
            temperatura *= tasa_enfriamiento

        return movimiento_actual

if __name__ == "__main__":
    ventana_principal = tk.Tk()
    # Hacemos que la ventana no se pueda redimensionar para mantener el diseño
    ventana_principal.resizable(False, False) 
    app = TatetiGUI(ventana_principal)
    ventana_principal.mainloop()
```


6. Diseñe e implemente un algoritmo genético para cargar una grúa con $n=10\;cajas$ que puede soportar un peso máximo $C=1000\;kg$. Cada caja *j* tiene asociado un precio $p_j$ y un peso $w_j$ como se indica en la tabla de abajo, de manera que el algoritmo debe ser capaz de maximizar el precio sin superar el límite de carga.

        6.1 En primer lugar, es necesario representar qué cajas estarán cargadas en la grúa y cuáles no. Esta representación corresponde a un Individuo con el que trabajará el algoritmo.

        6.2 A continuación, genere una Población que contenga un número $N$ de individuos (se recomienda elegir un número par). Es necesario crear un control que verifique que ninguno de los individuos supere el peso límite.

        6.3 Cree ahora una función que permita evaluar la Idoneidad de cada individuo y seleccione $N/2$ parejas usando el método de la ruleta.

        6.4 Por último, Cruce las parejas elegidas, aplique un mecanismo de Mutación y verifique que los individuos de la nueva población no superen el límite de peso.

        6.5 Realice este proceso iterativamente hasta que se cumpla el mecanismo de detención de su elección y muestre el mejor individuo obtenido junto con el peso y el precio que alcanza.

```python
"""
Algoritmo Genético para el Problema de la Mochila (Knapsack)
Trabajo Práctico - Inteligencia Artificial 1 - Ingeniería Mecatrónica

Carga de una grúa: n = 10 cajas, capacidad máxima C = 1000 kg.
Objetivo: maximizar el precio total sin superar el límite de peso.
"""

import random
from typing import List, Tuple

import matplotlib.pyplot as plt

# =================================================================
# 1. DATOS DEL PROBLEMA
# =================================================================
# Precio y peso de cada una de las 10 cajas (índice 0 = caja 1, ..., índice 9 = caja 10).
PRECIOS: List[int] = [100, 50, 115, 25, 200, 30, 40, 100, 100, 100]
PESOS: List[int] = [300, 200, 450, 145, 664, 90, 150, 355, 401, 395]

# Capacidad máxima de carga de la grúa (restricción del problema).
CAPACIDAD_MAXIMA: int = 1000

# Cantidad de genes que tendrá cada cromosoma (una posición por caja).
N_CAJAS: int = len(PRECIOS)


# =================================================================
# 2. REPRESENTACIÓN DEL INDIVIDUO (6.1)
# =================================================================
class Individuo:
    """
    Representa una solución candidata al problema de la mochila.

    El cromosoma es un arreglo binario de longitud N_CAJAS, donde
    genes[j] == 1 significa que la caja j se carga en la grúa, y
    genes[j] == 0 significa que no se carga.
    Ej: [1, 0, 0, 0, 1, 0, 0, 0, 0, 0] -> se cargan las cajas 1 y 5.
    """

    def __init__(self, genes: List[int]) -> None:
        # Lista de 0s y 1s: el "cromosoma" del individuo.
        self.genes: List[int] = genes

    def peso_total(self) -> int:
        """Suma los pesos únicamente de las cajas cargadas (gen == 1)."""
        # zip empareja cada gen con su peso correspondiente y filtramos
        # solo las posiciones donde el gen vale 1 (caja incluida).
        return sum(w for gen, w in zip(self.genes, PESOS) if gen == 1)

    def fitness(self) -> int:
        """
        Función de Aptitud (6.3): el fitness de un individuo es el
        precio total de las cajas que efectivamente carga.
        Cuanto mayor el fitness, mejor es la solución (es un problema
        de maximización).
        """
        # Misma lógica que peso_total() pero sumando precios en vez de pesos.
        return sum(p for gen, p in zip(self.genes, PRECIOS) if gen == 1)

    def copy(self) -> "Individuo":
        """Devuelve una copia independiente del individuo (nueva lista de genes)."""
        # Importante: .copy() en la lista para no compartir referencia y así
        # evitar que modificar el hijo afecte al padre original.
        return Individuo(self.genes.copy())

    def __repr__(self) -> str:
        # Representación legible útil para debug o impresión en consola.
        return (
            f"Individuo(genes={self.genes}, "
            f"peso={self.peso_total()}kg, precio={self.fitness()})"
        )


# =================================================================
# 3. ALGORITMO GENÉTICO
# =================================================================
class AlgoritmoGeneticoMochila:
    """
    Implementación de un Algoritmo Genético clásico (generacional) para
    resolver el problema de la mochila con reparación de individuos.
    """

    def __init__(
        self,
        tam_poblacion: int = 20,
        tasa_mutacion: float = 0.10,
        max_generaciones: int = 100,
        generaciones_sin_mejora_limite: int = 15,
    ) -> None:
        # Hiperparámetros del algoritmo (todos configurables desde afuera).
        self.tam_poblacion = tam_poblacion
        self.tasa_mutacion = tasa_mutacion
        self.max_generaciones = max_generaciones
        self.generaciones_sin_mejora_limite = generaciones_sin_mejora_limite

        # Lista donde se guarda el mejor fitness encontrado en cada
        # generación. Se usa después para graficar la curva de convergencia.
        self.historial_mejor_fitness: List[int] = []

    # -----------------------------------------------------------
    # 6.2 Población inicial (con control de peso)
    # -----------------------------------------------------------
    def _generar_individuo_valido(self) -> Individuo:
        """
        Genera un individuo aleatorio y lo reintenta hasta que su peso
        no supere la capacidad máxima de la grúa (1000 kg).
        """
        # Bucle "generar y verificar": se crean cromosomas al azar hasta
        # obtener uno que respete la restricción de peso. Al ser solo
        # 10 genes, esto converge muy rápido en la práctica.
        while True:
            genes = [random.randint(0, 1) for _ in range(N_CAJAS)]
            individuo = Individuo(genes)
            if individuo.peso_total() <= CAPACIDAD_MAXIMA:
                return individuo

    def generar_poblacion_inicial(self) -> List[Individuo]:
        """Genera una población de N individuos, todos con peso válido."""
        # Se repite la generación de un individuo válido tantas veces
        # como indique tam_poblacion (por defecto, N = 20).
        return [self._generar_individuo_valido() for _ in range(self.tam_poblacion)]

    # -----------------------------------------------------------
    # 6.3 Selección por ruleta
    # -----------------------------------------------------------
    def _seleccion_ruleta(self, poblacion: List[Individuo]) -> Individuo:
        """Selecciona un individuo mediante el método de la ruleta,
        proporcional a su fitness (precio total)."""
        # Fitness de cada individuo de la población: define su "porción"
        # de la ruleta (a mayor precio total, mayor probabilidad de ser elegido).
        fitnesses = [ind.fitness() for ind in poblacion]
        suma_total = sum(fitnesses)

        if suma_total == 0:
            # Caso degenerado: todos los individuos tienen fitness 0
            # (ninguna caja cargada). Se elige uno al azar para no dividir por 0.
            return random.choice(poblacion)

        # Se "tira la bolita" de la ruleta: un número aleatorio entre 0 y
        # la suma total de fitness.
        pick = random.uniform(0, suma_total)

        # Se recorre la población acumulando fitness hasta superar el valor
        # sorteado; el individuo donde esto ocurre es el seleccionado.
        acumulado = 0
        for individuo, fit in zip(poblacion, fitnesses):
            acumulado += fit
            if acumulado >= pick:
                return individuo

        return poblacion[-1]  # Salvaguarda por redondeo de flotantes.

    def seleccionar_parejas(
        self, poblacion: List[Individuo]
    ) -> List[Tuple[Individuo, Individuo]]:
        """Selecciona N/2 parejas de padres mediante ruleta."""
        # Con N=20 individuos se forman 10 parejas (cada una producirá 2 hijos,
        # manteniendo el tamaño de población constante entre generaciones).
        n_parejas = self.tam_poblacion // 2
        return [
            (self._seleccion_ruleta(poblacion), self._seleccion_ruleta(poblacion))
            for _ in range(n_parejas)
        ]

    # -----------------------------------------------------------
    # 6.4.1 Cruce de un punto
    # -----------------------------------------------------------
    def _cruce_un_punto(
        self, padre1: Individuo, padre2: Individuo
    ) -> Tuple[Individuo, Individuo]:
        """Aplica cruce de un solo punto entre dos padres, generando dos hijos."""
        # Se elige un índice de corte al azar entre 1 y N_CAJAS-1 (nunca en
        # el extremo 0, porque ahí no habría intercambio real de material genético).
        punto_corte = random.randint(1, N_CAJAS - 1)

        # Hijo 1: primera parte del padre1 + segunda parte del padre2.
        genes_hijo1 = padre1.genes[:punto_corte] + padre2.genes[punto_corte:]
        # Hijo 2: primera parte del padre2 + segunda parte del padre1 (complementario).
        genes_hijo2 = padre2.genes[:punto_corte] + padre1.genes[punto_corte:]

        return Individuo(genes_hijo1), Individuo(genes_hijo2)

    # -----------------------------------------------------------
    # 6.4.2 Mutación (bit-flip)
    # -----------------------------------------------------------
    def _mutar(self, individuo: Individuo) -> None:
        """Recorre cada gen del individuo y, con probabilidad tasa_mutacion,
        invierte su valor (0 -> 1, 1 -> 0)."""
        # Se recorre gen por gen (no el cromosoma entero de una vez), ya que
        # la mutación es independiente para cada posición del arreglo.
        for i in range(len(individuo.genes)):
            if random.random() < self.tasa_mutacion:
                # Bit-flip: si es 0 pasa a 1, si es 1 pasa a 0.
                individuo.genes[i] = 1 - individuo.genes[i]

    # -----------------------------------------------------------
    # 6.4.3 Validación y Reparación
    # -----------------------------------------------------------
    def _reparar(self, individuo: Individuo) -> None:
        """
        Si el individuo supera la capacidad máxima, elige aleatoriamente
        genes en estado 1 y los apaga (0) de manera iterativa hasta que
        el peso vuelva a ser válido (<= 1000 kg).
        """
        # Se repite mientras el individuo siga siendo inválido (peso excedido).
        while individuo.peso_total() > CAPACIDAD_MAXIMA:
            # Se identifican las posiciones de las cajas actualmente cargadas.
            indices_cargados = [i for i, gen in enumerate(individuo.genes) if gen == 1]
            if not indices_cargados:
                # Seguridad: si no queda ninguna caja cargada, el peso ya
                # sería 0 y el while no debería seguir; esto evita un bucle infinito.
                break
            # Se descarga una caja al azar (no la más pesada ni la más barata,
            # para no introducir un sesgo determinístico en la reparación).
            idx_a_descargar = random.choice(indices_cargados)
            individuo.genes[idx_a_descargar] = 0

    def generar_nueva_poblacion(
        self, parejas: List[Tuple[Individuo, Individuo]]
    ) -> List[Individuo]:
        """
        Para cada pareja de padres aplica, EN ORDEN ESTRICTO:
        1) Cruce de un punto -> 2) Mutación bit-flip -> 3) Reparación de peso.
        """
        nueva_poblacion: List[Individuo] = []
        for padre1, padre2 in parejas:
            # Paso 1: cruce -> se generan 2 hijos a partir de los 2 padres.
            hijo1, hijo2 = self._cruce_un_punto(padre1, padre2)

            for hijo in (hijo1, hijo2):
                # Paso 2: mutación -> se aplica bit-flip a cada hijo.
                self._mutar(hijo)
                # Paso 3: reparación -> se corrige el peso si quedó inválido.
                self._reparar(hijo)
                nueva_poblacion.append(hijo)

        return nueva_poblacion

    # -----------------------------------------------------------
    # 6.5 Ciclo evolutivo con Early Stopping
    # -----------------------------------------------------------
    def ejecutar(self) -> Individuo:
        """
        Ejecuta el ciclo evolutivo completo hasta max_generaciones,
        o hasta que el mejor fitness no mejore durante
        generaciones_sin_mejora_limite generaciones consecutivas.
        """
        # Generación 0: población inicial aleatoria (todos con peso válido).
        poblacion = self.generar_poblacion_inicial()

        # Se guarda el mejor individuo visto hasta el momento (elitismo "pasivo":
        # no se reinserta en la población, solo se recuerda para no perderlo).
        mejor_individuo = max(poblacion, key=lambda ind: ind.fitness()).copy()
        self.historial_mejor_fitness = [mejor_individuo.fitness()]

        # Contador de generaciones consecutivas sin mejora, para el early stopping.
        generaciones_sin_mejora = 0

        for generacion in range(1, self.max_generaciones + 1):
            # Selección de padres por ruleta (6.3).
            parejas = self.seleccionar_parejas(poblacion)
            # Cruce + mutación + reparación, en orden estricto (6.4).
            poblacion = self.generar_nueva_poblacion(parejas)

            # Se busca el mejor individuo de la generación recién creada.
            mejor_de_la_generacion = max(poblacion, key=lambda ind: ind.fitness())

            if mejor_de_la_generacion.fitness() > mejor_individuo.fitness():
                # Hubo mejora: se actualiza el mejor histórico y se reinicia el contador.
                mejor_individuo = mejor_de_la_generacion.copy()
                generaciones_sin_mejora = 0
            else:
                # No hubo mejora en esta generación: se incrementa el contador.
                generaciones_sin_mejora += 1

            # Se registra el mejor fitness histórico (no el de la generación actual)
            # para que la curva de convergencia sea siempre no decreciente.
            self.historial_mejor_fitness.append(mejor_individuo.fitness())

            # Early Stopping: si pasaron demasiadas generaciones sin mejora, se corta.
            if generaciones_sin_mejora >= self.generaciones_sin_mejora_limite:
                print(
                    f"Early stopping en la generación {generacion}: "
                    f"sin mejora durante {self.generaciones_sin_mejora_limite} "
                    f"generaciones consecutivas."
                )
                break

        return mejor_individuo

    def graficar_convergencia(self) -> None:
        """Grafica la evolución del mejor fitness a lo largo de las generaciones."""
        plt.figure(figsize=(10, 5))
        # historial_mejor_fitness contiene un valor por cada generación
        # (incluida la generación 0, la población inicial).
        plt.plot(
            self.historial_mejor_fitness,
            marker="o",
            markersize=3,
            linewidth=1.5,
            color="#2E86AB",
        )
        plt.title("Curva de Convergencia - Algoritmo Genético (Knapsack)")
        plt.xlabel("Generación")
        plt.ylabel("Mejor Fitness (Precio Total)")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()


# =================================================================
# 4. BLOQUE DE PRUEBA / EJECUCIÓN
# =================================================================
if __name__ == "__main__":
    # Se instancia el algoritmo con los parámetros pedidos por la consigna:
    # N=20, mutación 10%, máximo 100 generaciones, early stopping a las 15
    # generaciones sin mejora.
    ga = AlgoritmoGeneticoMochila(
        tam_poblacion=20,
        tasa_mutacion=0.10,
        max_generaciones=100,
        generaciones_sin_mejora_limite=15,
    )

    # Corre el ciclo evolutivo completo y devuelve el mejor individuo encontrado.
    mejor_solucion = ga.ejecutar()

    # Muestra la curva de convergencia (Fitness vs. Generación).
    ga.graficar_convergencia()

    # Imprime el resultado final: cromosoma, peso y precio total.
    print("\n=== RESULTADO FINAL ===")
    print(f"Cromosoma ganador: {mejor_solucion.genes}")
    print(f"Peso total: {mejor_solucion.peso_total()} kg (límite: {CAPACIDAD_MAXIMA} kg)")
    print(f"Precio total (fitness): {mejor_solucion.fitness()}")
    # Se traducen los índices (0..9) a números de caja (1..10) para mayor claridad.
    cajas_cargadas = [j + 1 for j, gen in enumerate(mejor_solucion.genes) if gen == 1]
    print(f"Cajas cargadas: {cajas_cargadas}")
```
