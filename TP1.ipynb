8. La Hormiga de Langton es un agente capaz de modificar el estado de la casilla en la que se encuentra para colorearla o bien de blanco o de negro. Al comenzar, la ubicación de la hormiga es una casilla aleatoria y mira hacia una de las cuatro casillas adyacentes. Si...

* ... la casilla sobre la que está es blanca, cambia el color del cuadrado, gira noventa grados a la derecha y avanza un cuadrado.

* ... la casilla sobre la que está es negra, cambia el color del cuadrado, gira noventa grados a la izquierda y avanza un cuadrado.

    Caracterice el agente con su tabla REAS y las propiedades del entorno para después programarlo en Python:

    ¿Observa que se repite algún patrón? De ser así, ¿a partir de qué iteración?

| Agente   | Medidas de Rendimiento | Etorno | Actuadores | Sensores |
|----------|------------------------|--------|------------|----------|
|Hormiga de Langton|Cantidades de avenidas que genera|Agrupacion de celdas|Cambio de celda|Codigo que verfica ña ceñda|

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML  # <-- NUEVO: Importación necesaria para Jupyter

#variables de configuración

BOARD_SIZE = 100  # definte tamaño de la cuadrícula
ANIMATION_SPEED = 50   # tiempo de espera entre cada fotograma en milisegundos
STEPS_PER_FRAME = 40   # aumentado para que el video muestre un avance rápido (La hormiga creará la "autopista" más rápido)
TOTAL_FRAMES = 300     # <-- NUEVO: En Jupyter es mejor definir un límite de frames para renderizar el video

BLACK = 0
WHITE = 1

COLOR_BLACK = (0.0, 0.0, 0.0)
COLOR_WHITE = (1.0, 1.0, 1.0)
COLOR_ANT = (1.0, 0.0, 0.0)

DIRECTIONS = [(-1, 0), (0, 1), (1, 0), (0, -1)]


class Board:
    def __init__(self, size):
        self.size = size
        self.grid = [[BLACK for _ in range(size)] for _ in range(size)]

    def is_inside(self, row, col):
        return 0 <= row < self.size and 0 <= col < self.size

    def get_color(self, row, col):
        return self.grid[row][col]

    def toggle_color(self, row, col):
        self.grid[row][col] = WHITE if self.grid[row][col] == BLACK else BLACK

    def to_rgb_array(self):
        array = np.zeros((self.size, self.size, 3))
        for row in range(self.size):
            for col in range(self.size):
                array[row][col] = COLOR_WHITE if self.grid[row][col] == WHITE else COLOR_BLACK
        return array


class Ant:
    def __init__(self, board, row, col, direction_index=0):
        self.board = board       
        self.row = row           
        self.col = col           
        self.direction_index = direction_index  
        self.alive = True        

    def step(self):
        if not self.alive:
            return

        if not self.board.is_inside(self.row, self.col):
            self.alive = False
            return

        current_color = self.board.get_color(self.row, self.col)

        if current_color == WHITE:
            self.board.toggle_color(self.row, self.col)  
            self.turn_right()                            
        else:
            self.board.toggle_color(self.row, self.col)  
            self.turn_left()                             

        self.move_forward()

    def turn_right(self):
        self.direction_index = (self.direction_index + 1) % 4

    def turn_left(self):
        self.direction_index = (self.direction_index - 1) % 4

    def move_forward(self):
        d_row, d_col = DIRECTIONS[self.direction_index]
        self.row += d_row
        self.col += d_col


def run_simulation():
    board = Board(BOARD_SIZE)
    center = BOARD_SIZE // 2 
    ant = Ant(board, center, center, direction_index=0)  

    iteration = 0 

    fig, ax = plt.subplots(figsize=(6, 6)) 
    im = ax.imshow(board.to_rgb_array())   
    title = ax.set_title(f"Hormiga de Langton — Iteración: {iteration}") 
    ax.axis("off") 

    def render():
        array = board.to_rgb_array()
        if ant.alive and board.is_inside(ant.row, ant.col):
            array[ant.row][ant.col] = COLOR_ANT
        im.set_data(array) 
        title.set_text(f"Hormiga de Langton — Iteración: {iteration}") 

    def update(frame):
        nonlocal iteration 
        if ant.alive:
            for _ in range(STEPS_PER_FRAME):
                was_inside = board.is_inside(ant.row, ant.col)
                ant.step() 
                iteration += 1
                
                if not ant.alive and was_inside:
                    pass # En Jupyter evitamos imprimir demasiados print() que ensucien la celda
                elif not ant.alive:
                    anim.event_source.stop() 
                    break
            render()
        return [im, title] 

    # <-- NUEVO: Agregamos 'frames=TOTAL_FRAMES' para que construya el video y termine
    anim = animation.FuncAnimation(
        fig, update, frames=TOTAL_FRAMES, interval=ANIMATION_SPEED, blit=False, cache_frame_data=False
    )
    
    # <-- NUEVO: Previene que Matplotlib dibuje una imagen estática duplicada debajo de la animación
    plt.close() 
    
    # <-- NUEVO: Devuelve el objeto HTML para que Jupyter renderice el reproductor interactivo
    return HTML(anim.to_jshtml())

# Ejecutamos la función y dejamos que Jupyter muestre el resultado en la celda
run_simulation()
```

9. El Juego de la Vida de Conway consiste en un tablero donde cada casilla representa una célula, de manera que a cada célula le rodean 8 vecinas. Las células tienen dos estados: están *vivas* o *muertas*. En cada iteración, el estado de todas las células se tiene en cuenta para calcular el estado siguiente en simultáneo de acuerdo a las siguientes acciones:

* Nacer: Si una célula muerta tiene exactamente 3 células vecinas vivas, dicha célula pasa a estar viva.

* Morir: Una célula viva puede morir sobrepoblación cuando tiene más de tres vecinos alrededor o por aislamiento si tiene solo un vecino o ninguno.

* Vivir: una célula se mantiene viva si tiene 2 o 3 vecinos a su alrededor.

    Caracterice el agente con su tabla REAS y las propiedades del entorno para después programarlo en Python:

| Agente   | Medidas de Rendimiento | Etorno | Actuadores | Sensores |
|----------|------------------------|--------|------------|----------|
|Célula|Maximización de la vida|Tablero 2D de casillas con 8 vecinas contiguas|Cambio de estado|Detector de estado de las 8 células vecinas|

```python
# =============================================================================
# IMPORTACIONES Y CONFIGURACIÓN INICIAL
# =============================================================================

from __future__ import annotations

# 'dataclass' y 'field' simplifican la creación de clases orientadas a datos.
from dataclasses import dataclass, field

# 'Final' es un indicador de tipo para definir constantes inmutables.
from typing import Final

# NumPy nos permite realizar operaciones matriciales vectorizadas extremadamente rápidas en C.
import numpy as np
from numpy.typing import NDArray

# Pygame es la librería gráfica encargada de renderizar la ventana emergente e interactuar con teclado/mouse.
import pygame


# =============================================================================
# 1. CONSTANTES DEL JUEGO Y PALETA DE COLORES
# =============================================================================

# Estados posibles de las celdas (usamos enteros simples para optimizar memoria)[cite: 1]
CELULA_VIVA: Final[int] = 1   # Representa una celda activa/viva[cite: 1]
CELULA_MUERTA: Final[int] = 0  # Representa una celda inactiva/muerta[cite: 1]

# Definición de la paleta visual en formato RGB (Rojo, Verde, Azul)[cite: 2]
COLOR_FONDO: Final[tuple[int, int, int]] = (18, 18, 18)          # Gris muy oscuro para el fondo general[cite: 2]
COLOR_CELULA_VIVA: Final[tuple[int, int, int]] = (0, 220, 130)   # Verde neón vibrante para las células[cite: 2]
COLOR_LINEA_CUADRICULA: Final[tuple[int, int, int]] = (40, 40, 40) # Gris tenue para separar celdas[cite: 2]
COLOR_TEXTO: Final[tuple[int, int, int]] = (230, 230, 230)      # Blanco suave para el texto de la interfaz[cite: 2]
COLOR_BARRA_SUPERIOR: Final[tuple[int, int, int]] = (28, 28, 28) # Gris medio para el panel de control[cite: 2]


# =============================================================================
# 2. ESTRUCTURA DE DATOS DEL TABLERO (ESTADO)
# =============================================================================

@dataclass
class Board:
    """Clase encargada ÚNICAMENTE de mantener la matriz de estado del tablero[cite: 1].
    
    Aplica el principio de responsabilidad única (SOLID): no sabe cómo se dibujan
    las células ni cuáles son las reglas de evolución[cite: 1].
    """
    alto: int   # Cantidad de filas horizontales del tablero[cite: 1]
    ancho: int  # Cantidad de columnas verticales del tablero[cite: 1]
    
    # Matriz NumPy 2D que contendrá los 0s y 1s.
    # 'init=False' indica que se generará automáticamente en __post_init__[cite: 1].
    celulas: NDArray[np.int8] = field(init=False)

    def __post_init__(self) -> None:
        """Inicializa el tablero lleno de ceros (todas las células muertas)[cite: 1]."""
        # Usamos dtype=np.int8 para consumir solo 1 byte por celda en memoria[cite: 1].
        self.celulas = np.zeros((self.alto, self.ancho), dtype=np.int8)

    def colocar_patron(
        self,
        patron: NDArray[np.int8],
        fila_superior_izquierda: int,
        columna_superior_izquierda: int,
    ) -> None:
        """Incrusta una sub-matriz (patrón predefinido) en una posición específica[cite: 1]."""
        alto_patron, ancho_patron = patron.shape
        fila_final = fila_superior_izquierda + alto_patron
        columna_final = columna_superior_izquierda + ancho_patron

        # Validación previa para evitar salirnos de los bordes del tablero[cite: 1]
        if fila_final > self.alto or columna_final > self.ancho:
            raise ValueError("El patrón excede los límites del tablero en esa posición.")

        # Reemplazamos la porción correspondiente de la matriz mediante 'slicing'[cite: 1]
        self.celulas[
            fila_superior_izquierda:fila_final,
            columna_superior_izquierda:columna_final,
        ] = patron

    def contar_celulas_vivas(self) -> int:
        """Suma de manera vectorizada el total de '1s' presentes en la matriz[cite: 1]."""
        return int(np.sum(self.celulas))


# =============================================================================
# 3. MOTOR DE LÓGICA DE SIMULACIÓN (REGLAS DE CONWAY)
# =============================================================================

class GameOfLife:
    """Clase encargada de calcular la evolución biológica del autómata[cite: 1]."""
    
    # Coordenadas relativas (delta_fila, delta_columna) de las 8 células vecinas
    # que rodean a cualquier celda central (Vecindad de Moore)[cite: 1].
    _DESPLAZAMIENTOS_VECINDAD: tuple[tuple[int, int], ...] = (
        (-1, -1), (-1, 0), (-1, 1),  # Fila superior (Izquierda, Centro, Derecha)
        (0, -1),           (0, 1),   # Misma fila (Izquierda, Derecha)
        (1, -1),  (1, 0),  (1, 1),   # Fila inferior (Izquierda, Centro, Derecha)
    )

    def __init__(self, tablero: Board) -> None:
        """Asocia el motor de juego al tablero actual[cite: 1]."""
        self.tablero = tablero

    def _contar_vecinas_vivas(self) -> NDArray[np.int16]:
        """Calcula una matriz donde cada posición indica cuántas vecinas vivas tiene esa celda[cite: 1].
        
        TÉCNICA DE RENDIMIENTO:
        En lugar de usar bucles 'for' celda por celda (lentos en Python),
        desplazamos la matriz entera en las 8 direcciones con 'np.roll'[cite: 1].
        Al sumar las 8 matrices desplazadas, obtenemos el total de vecinas[cite: 1].
        Además, 'np.roll' implementa de forma matemática una topología toroidal
        (los bordes se conectan: el borde izquierdo toca al derecho y el superior al inferior)[cite: 1].
        """
        conteo_vecinas = np.zeros_like(self.tablero.celulas, dtype=np.int16)

        for delta_fila, delta_columna in self._DESPLAZAMIENTOS_VECINDAD:
            # np.roll rota circularmente la matriz entera[cite: 1]
            tablero_desplazado = np.roll(
                self.tablero.celulas,
                shift=(delta_fila, delta_columna),
                axis=(0, 1),
            )
            conteo_vecinas += tablero_desplazado

        return conteo_vecinas

    def avanzar_una_generacion(self) -> None:
        """Aplica simultáneamente las 4 reglas del Juego de la Vida de Conway[cite: 1]:
        
        1. Supervivencia: Célula viva con 2 o 3 vecinas vivas permanece VIVA[cite: 1].
        2. Muerte por Aislamiento: Célula viva con menos de 2 vecinas MUERE[cite: 1].
        3. Muerte por Sobrepoblación: Célula viva con más de 3 vecinas MUERE[cite: 1].
        4. Nacimiento: Célula muerta con exactamente 3 vecinas NACIMIENTO (pasa a VIVA)[cite: 1].
        """
        estado_actual = self.tablero.celulas
        conteo_vecinas_vivas = self._contar_vecinas_vivas()

        # Máscaras booleanas de evaluación[cite: 1]
        celulas_vivas = estado_actual == CELULA_VIVA
        
        # Una célula viva sobrevive si tiene 2 o 3 vecinas[cite: 1]
        sobrevive = celulas_vivas & np.isin(conteo_vecinas_vivas, [2, 3])
        
        # Una célula muerta nace si tiene exactamente 3 vecinas[cite: 1]
        nace = (~celulas_vivas) & (conteo_vecinas_vivas == 3)

        # Construcción atómica del nuevo estado (0 o 1)[cite: 1]
        self.tablero.celulas = np.where(
            sobrevive | nace, CELULA_VIVA, CELULA_MUERTA
        ).astype(np.int8)


# =============================================================================
# 4. PATRONES PREDEFINIDOS (FORMAS CLÁSICAS)
# =============================================================================

class PatronesClasicos:
    """Contenedor de arreglos bidimensionales con figuras conocidas[cite: 1]."""
    
    @staticmethod
    def planeador() -> NDArray[np.int8]:
        """Glider: Pequeña nave espacial que viaja diagonalmente por el tablero[cite: 1]."""
        return np.array([
            [0, 1, 0],
            [0, 0, 1],
            [1, 1, 1],
        ], dtype=np.int8)

    @staticmethod
    def pulsar() -> NDArray[np.int8]:
        """Pulsar: Oscilador complejo con período de 3 generaciones[cite: 1]."""
        pulsar = np.zeros((13, 13), dtype=np.int8)
        segmentos_fila = [2, 3, 4, 8, 9, 10]
        segmentos_columna = [0, 5, 7, 12]

        # Construcción mediante simetría de cuadrantes[cite: 1]
        for col in segmentos_columna:
            for fila in segmentos_fila:
                pulsar[fila, col] = CELULA_VIVA

        for fila in segmentos_columna:
            for col in segmentos_fila:
                pulsar[fila, col] = CELULA_VIVA

        return pulsar


# =============================================================================
# 5. RENDERIZADOR GRÁFICO (INTERFAZ CON PYGAME)
# =============================================================================

class RenderizadorPygame:
    """Gestiona la apertura, renderizado e ilustración gráfica en la ventana[cite: 2]."""

    def __init__(
        self,
        tablero: Board,
        tamano_celda_en_pixeles: int = 16,
        alto_barra_superior: int = 40,
    ) -> None:
        """Calcula el tamaño necesario de la ventana e inicializa Pygame[cite: 2]."""
        self.tamano_celda_en_pixeles = tamano_celda_en_pixeles
        self.alto_barra_superior = alto_barra_superior

        # Cálculo en píxeles de las dimensiones reales de la ventana[cite: 2]
        ancho_ventana = tablero.ancho * tamano_celda_en_pixeles
        alto_ventana = tablero.alto * tamano_celda_en_pixeles + alto_barra_superior

        # Inicialización formal de subsistemas gráficos y de fuentes[cite: 2]
        pygame.display.init()
        pygame.font.init()

        # Creación de la ventana contenedora nativa del sistema operativo[cite: 2]
        self.superficie_ventana = pygame.display.set_mode((ancho_ventana, alto_ventana))
        pygame.display.set_caption("El Juego de la Vida de Conway")
        
        # Carga de la tipografía para los datos superiores[cite: 2]
        self._fuente_texto = pygame.font.SysFont("consolas", 16)

    def renderizar(
        self,
        tablero: Board,
        numero_generacion: int,
        simulacion_en_pausa: bool = False,
    ) -> None:
        """Dibuja el fotograma actual ordenadamente y lo envía a la pantalla[cite: 2]."""
        # 1. Limpia toda la pantalla pintándola del color de fondo[cite: 2]
        self.superficie_ventana.fill(COLOR_FONDO)

        # 2. Dibuja las células vivas[cite: 2]
        self._dibujar_celulas(tablero)

        # 3. Dibuja la rejilla del tablero[cite: 2]
        self._dibujar_lineas_de_cuadricula(tablero)

        # 4. Dibuja la barra de información superior[cite: 2]
        self._dibujar_barra_de_informacion(tablero, numero_generacion, simulacion_en_pausa)

        # 5. Intercambia el buffer interno con el visible para evitar parpadeos[cite: 2]
        pygame.display.flip()

    def _dibujar_celulas(self, tablero: Board) -> None:
        """Localiza e ilustra únicamente las celdas en estado VIVO[cite: 2].
        
        OPTIMIZACIÓN: 'np.argwhere' nos da directamente las coordenadas (fila, col)
        donde la celda vale 1. No perdemos tiempo iterando sobre celdas muertas[cite: 2].
        """
        coordenadas_vivas = np.argwhere(tablero.celulas == CELULA_VIVA)
        for fila, columna in coordenadas_vivas:
            # Convertimos la posición matricial a píxeles de la ventana[cite: 2]
            rectangulo_celula = pygame.Rect(
                columna * self.tamano_celda_en_pixeles,
                fila * self.tamano_celda_en_pixeles + self.alto_barra_superior,
                self.tamano_celda_en_pixeles,
                self.tamano_celda_en_pixeles,
            )
            pygame.draw.rect(self.superficie_ventana, COLOR_CELULA_VIVA, rectangulo_celula)

    def _dibujar_lineas_de_cuadricula(self, tablero: Board) -> None:
        """Taza las líneas verticales y horizontales de la cuadrícula[cite: 2]."""
        ancho_total = tablero.ancho * self.tamano_celda_en_pixeles
        alto_total = tablero.alto * self.tamano_celda_en_pixeles

        # Líneas verticales[cite: 2]
        for col in range(tablero.ancho + 1):
            pos_x = col * self.tamano_celda_en_pixeles
            pygame.draw.line(
                self.superficie_ventana,
                COLOR_LINEA_CUADRICULA,
                (pos_x, self.alto_barra_superior),
                (pos_x, self.alto_barra_superior + alto_total),
            )

        # Líneas horizontales[cite: 2]
        for fila in range(tablero.alto + 1):
            pos_y = fila * self.tamano_celda_en_pixeles + self.alto_barra_superior
            pygame.draw.line(
                self.superficie_ventana,
                COLOR_LINEA_CUADRICULA,
                (0, pos_y),
                (ancho_total, pos_y),
            )

    def _dibujar_barra_de_informacion(
        self,
        tablero: Board,
        numero_generacion: int,
        simulacion_en_pausa: bool,
    ) -> None:
        """Pinta la franja horizontal superior con datos y controles[cite: 2]."""
        ancho_total = tablero.ancho * self.tamano_celda_en_pixeles
        rect_barra = pygame.Rect(0, 0, ancho_total, self.alto_barra_superior)
        pygame.draw.rect(self.superficie_ventana, COLOR_BARRA_SUPERIOR, rect_barra)

        # Texto informativo interactivo[cite: 2]
        estado_texto = "PAUSA" if simulacion_en_pausa else "EJECUTANDO"
        texto = (
            f"Gen: {numero_generacion} | Vivas: {tablero.contar_celulas_vivas()} "
            f"| [{estado_texto}] (ESPACIO=Pausa · R=Reiniciar · Click=Dibujar · ESC=Salir)"
        )
        superficie_texto = self._fuente_texto.render(texto, True, COLOR_TEXTO)
        self.superficie_ventana.blit(superficie_texto, (10, 10))

    def cerrar(self) -> None:
        """Libera de memoria los recursos gráficos consumidos[cite: 2]."""
        pygame.font.quit()
        pygame.display.quit()


# =============================================================================
# 6. APLICACIÓN PRINCIPAL (BUCLE DE EVENTOS Y CONTROL)
# =============================================================================

class AplicacionJuegoDeLaVidaPygame:
    """Orquesta la ejecución interactiva: teclado, ratón y tasa de refresco[cite: 2]."""

    def __init__(
        self,
        tablero: Board,
        renderizador: RenderizadorPygame,
        constructor_tablero_inicial,
        cuadros_por_segundo: int = 10,
    ) -> None:
        """Configura los controladores esenciales del bucle de juego[cite: 2]."""
        self.tablero = tablero
        self.motor_de_juego = GameOfLife(tablero)
        self.renderizador = renderizador
        self._constructor_tablero_inicial = constructor_tablero_inicial
        self.cuadros_por_segundo = cuadros_por_segundo

        # Reloj interno de Pygame para fijar la velocidad máxima (FPS)[cite: 2]
        self._reloj = pygame.time.Clock()
        self._numero_generacion = 0
        self._simulacion_en_pausa = False

    def ejecutar(self) -> None:
        """Bucle infinito principal: procesa entrada, evoluciona y renderiza[cite: 2]."""
        activa = True
        while activa:
            # 1. Atiende acciones del usuario (Mouse/Teclado/Cerrar)[cite: 2]
            activa = self._procesar_eventos()

            # 2. Si el juego no está pausado, avanza 1 generación[cite: 2]
            if not self._simulacion_en_pausa:
                self.motor_de_juego.avanzar_una_generacion()
                self._numero_generacion += 1

            # 3. Redibuja los cambios[cite: 2]
            self.renderizador.renderizar(
                self.tablero,
                self._numero_generacion,
                self._simulacion_en_pausa,
            )

            # 4. Frena la ejecución para no saturar el CPU (limita los FPS)[cite: 2]
            self._reloj.tick(self.cuadros_por_segundo)

        # Cuando el bucle termina, cierra ordenadamente la ventana[cite: 2]
        self.renderizador.cerrar()

    def _procesar_eventos(self) -> bool:
        """Captura los clics e interacciones de la cola de eventos de Pygame[cite: 2]."""
        for evento in pygame.event.get():
            # Evento: Clic en la "X" de cerrar la ventana[cite: 2]
            if evento.type == pygame.QUIT:
                return False

            # Eventos de Teclado[cite: 2]
            if evento.type == pygame.KEYDOWN:
                if evento.key == pygame.K_ESCAPE:  # Tecla ESC -> Salir[cite: 2]
                    return False
                if evento.key == pygame.K_SPACE:   # Tecla ESPACIO -> Pausar/Reanudar[cite: 2]
                    self._simulacion_en_pausa = not self._simulacion_en_pausa
                if evento.key == pygame.K_r:       # Tecla R -> Reiniciar simulación[cite: 2]
                    self._reiniciar()

            # Eventos de Ratón (Click Izquierdo = Botón 1)[cite: 2]
            if evento.type == pygame.MOUSEBUTTONDOWN and evento.button == 1:
                # Solo permitimos pintar células si la simulación está pausada[cite: 2]
                if self._simulacion_en_pausa:
                    self._alternar_celula_click(evento.pos)

        return True

    def _alternar_celula_click(self, pos: tuple[int, int]) -> None:
        """Mapea las coordenadas en píxeles del clic hacia la matriz del tablero[cite: 2]."""
        x, y = pos
        y_relativa = y - self.renderizador.alto_barra_superior

        # Ignoramos clics sobre la barra de menú superior[cite: 2]
        if y_relativa < 0:
            return

        # División entera '//' para saber en qué fila/columna exacta se hizo clic[cite: 2]
        col = x // self.renderizador.tamano_celda_en_pixeles
        fila = y_relativa // self.renderizador.tamano_celda_en_pixeles

        # Si el clic fue dentro de los límites, invertimos el estado con XOR bit a bit (^= 1)[cite: 2]
        if 0 <= fila < self.tablero.alto and 0 <= col < self.tablero.ancho:
            self.tablero.celulas[fila, col] ^= 1  # Si era 0 pasa a 1; si era 1 pasa a 0[cite: 2]

    def _reiniciar(self) -> None:
        """Reconstruye el tablero original de demostración[cite: 2]."""
        tablero_nuevo = self._constructor_tablero_inicial()
        self.tablero.celulas = tablero_nuevo.celulas
        self.motor_de_juego = GameOfLife(self.tablero)
        self._numero_generacion = 0


# =============================================================================
# 7. INICIALIZADOR Y EJECUCIÓN DIRECTA
# =============================================================================

def construir_tablero_de_demostracion() -> Board:
    """Crea una instancia de tablero rellena con figuras iniciales[cite: 1, 2]."""
    tablero = Board(alto=30, ancho=50)
    # Colocamos un Planeador en la esquina superior izquierda[cite: 1, 2]
    tablero.colocar_patron(PatronesClasicos.planeador(), fila_superior_izquierda=1, columna_superior_izquierda=1)
    # Colocamos un Pulsar hacia el centro-derecho[cite: 1, 2]
    tablero.colocar_patron(PatronesClasicos.pulsar(), fila_superior_izquierda=10, columna_superior_izquierda=25)
    return tablero


def ejecutar_juego_de_la_vida(
    tamano_celda_en_pixeles: int = 16,
    cuadros_por_segundo: int = 10,
) -> None:
    """Punto de entrada diseñado para invocarse en una celda de Jupyter[cite: 2]."""
    pygame.init()
    
    # BLOQUE 'TRY... FINALLY':
    # Es fundamental en Jupyter Notebook. Garantiza que 'pygame.quit()' se ejecute
    # SIEMPRE que se cierre la ventana o haya un error, evitando que el Kernel de Jupyter
    # se quede "bloqueado" o "colgado"[cite: 2].
    try:
        tablero_inicial = construir_tablero_de_demostracion()
        renderizador = RenderizadorPygame(
            tablero_inicial,
            tamano_celda_en_pixeles=tamano_celda_en_pixeles
        )
        aplicacion = AplicacionJuegoDeLaVidaPygame(
            tablero=tablero_inicial,
            renderizador=renderizador,
            constructor_tablero_inicial=construir_tablero_de_demostracion,
            cuadros_por_segundo=cuadros_por_segundo,
        )
        # Inicia el bucle de la aplicación[cite: 2]
        aplicacion.ejecutar()
    finally:
        # Destruye la ventana y libera la consola de Jupyter al finalizar[cite: 2]
        pygame.quit()


# Al ejecutar esta celda, se llama automáticamente a la función principal
ejecutar_juego_de_la_vida()
```
