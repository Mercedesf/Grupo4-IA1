```python
import tkinter as tk

# 1. Definición de las coordenadas de las casillas en la cuadrícula (fila, columna)
coords = {
    'A': (0, 3), 'B': (0, 4),
    'C': (1, 3), 'D': (1, 4), 'E': (1, 5),
    'G': (2, 0), 'I': (2, 1), 'W': (2, 2), 'K': (2, 3), 'M': (2, 4), 'N': (2, 5),
    'P': (3, 0), 'Q': (3, 1), 'R': (3, 2), 'T': (3, 3), 'F': (3, 4)
}

# 2. Conexiones adyacentes respetando las paredes rojas y límites del tablero
vecinos = {
    'A': ['B', 'C'],
    'B': ['A'],          
    'C': ['A', 'K'],     
    'D': ['M'],          
    'E': ['N'],          
    'G': ['I', 'P'],
    'I': ['G', 'Q', 'W'],
    'K': ['C', 'M', 'T', 'W'],
    'M': ['D', 'K', 'N', 'F'],
    'N': ['E', 'M'],
    'P': ['G', 'Q'],
    'Q': ['I', 'P', 'R'],
    'R': ['Q', 'T'],     
    'T': ['K', 'R'],     
    'F': ['M'],          
    'W': ['I', 'K']      
}

# 3. Algoritmo de Búsqueda Primero en Profundidad (DFS)
def dfs(inicio, objetivo):
    stack = [(inicio, [inicio])]
    visitados = []
    camino_final = []
    
    while stack:
        nodo, camino = stack.pop()
        
        if nodo not in visitados:
            visitados.append(nodo)
            
            if nodo == objetivo:
                camino_final = camino
                break
                
            sucesores = sorted([v for v in vecinos[nodo] if v not in visitados], reverse=True)
            for v in sucesores:
                stack.append((v, camino + [v]))
                
    return visitados, camino_final

# Ejecutamos el algoritmo
inicio = 'I'
objetivo = 'F'
visitados, camino_optimo = dfs(inicio, objetivo)

# --- INICIO DE LA INTERFAZ GRÁFICA ---

TAMANO_CELDA = 80
FILAS = 4
COLUMNAS = 6

COLOR_FONDO = '#FFFFFF'
COLOR_INICIO_FIN = '#FFEB3B' # Amarillo inicial
COLOR_VISITADO = '#BBDEFB'   # Azul tenue
COLOR_CAMINO = '#4CAF50'     # Verde fuerte (se aplicará a todo el camino)
COLOR_PARED = '#D32F2F'

ventana = tk.Tk()
ventana.title("Animación DFS - Búsqueda de IA")
ventana.resizable(False, False)

canvas = tk.Canvas(ventana, width=COLUMNAS * TAMANO_CELDA, height=FILAS * TAMANO_CELDA, bg='white')
canvas.pack(padx=20, pady=20)

rectangulos = {}

# 4. Dibujar la cuadrícula base
for letra, (f, c) in coords.items():
    x1 = c * TAMANO_CELDA
    y1 = f * TAMANO_CELDA
    x2 = x1 + TAMANO_CELDA
    y2 = y1 + TAMANO_CELDA
    
    color_actual = COLOR_INICIO_FIN if letra in (inicio, objetivo) else COLOR_FONDO
    
    rect_id = canvas.create_rectangle(x1, y1, x2, y2, fill=color_actual, outline='#757575', width=2)
    rectangulos[letra] = rect_id
    
    canvas.create_text(x1 + TAMANO_CELDA/2, y1 + TAMANO_CELDA/2, text=letra, font=('Arial', 18, 'bold'))

# 5. Dibujar las Paredes
paredes_logicas = [
    ((1, 3), (1, 4)),
    ((1, 4), (1, 5)),
    ((2, 2), (3, 2)),
    ((3, 3), (3, 4))
]

for (f1, c1), (f2, c2) in paredes_logicas:
    if f1 == f2:
        x = max(c1, c2) * TAMANO_CELDA
        y_inicio = f1 * TAMANO_CELDA
        y_fin = y_inicio + TAMANO_CELDA
        canvas.create_line(x, y_inicio, x, y_fin, fill=COLOR_PARED, width=6)
    else:
        y = max(f1, f2) * TAMANO_CELDA
        x_inicio = c1 * TAMANO_CELDA
        x_fin = x_inicio + TAMANO_CELDA
        canvas.create_line(x_inicio, y, x_fin, y, fill=COLOR_PARED, width=6)

# 6. Funciones de Animación
def animar_busqueda(indice=0):
    if indice < len(visitados):
        nodo_actual = visitados[indice]
        # Durante la exploración, mantenemos el color amarillo del inicio y fin
        if nodo_actual not in (inicio, objetivo):
            canvas.itemconfig(rectangulos[nodo_actual], fill=COLOR_VISITADO) 
        
        ventana.after(400, animar_busqueda, indice + 1)
    else:
        ventana.after(500, animar_camino, 0)

def animar_camino(indice=0):
    if indice < len(camino_optimo):
        nodo_actual = camino_optimo[indice]
        
        # MODIFICACIÓN: Ahora aplicamos el color del camino a TODAS las celdas de la ruta (incluyendo I y F)
        canvas.itemconfig(rectangulos[nodo_actual], fill=COLOR_CAMINO) 
            
        ventana.after(300, animar_camino, indice + 1)

boton_iniciar = tk.Button(ventana, text="Iniciar Animación de Búsqueda", font=('Arial', 12, 'bold'), 
                          command=lambda: animar_busqueda(0))
boton_iniciar.pack(pady=10)

ventana.mainloop()
```

### Implementación 5.3 ###

```python
import tkinter as tk
import heapq

# --- 1. CONFIGURACIÓN DEL MAPA ---

# Coordenadas según la imagen (fila, columna)
coords = {
    'A': (0, 3), 'B': (0, 4),
    'C': (1, 3), 'D': (1, 4), 'E': (1, 5),
    'G': (2, 0), 'I': (2, 1), 'W': (2, 2), 'K': (2, 3), 'M': (2, 4), 'N': (2, 5),
    'P': (3, 0), 'Q': (3, 1), 'R': (3, 2), 'T': (3, 3), 'F': (3, 4)
}

# Conexiones adyacentes reales (Respetando las paredes rojas de la imagen)
# Paredes: C-D, D-E, W-R, T-F
vecinos = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],     # Sin pared con D según la imagen
    'C': ['A', 'K'],     # Pared con D
    'D': ['B', 'M'],     # Pared con C y E
    'E': ['N'],          # Pared con D
    'G': ['I', 'P'],
    'I': ['G', 'W', 'Q'],
    'W': ['I', 'K'],     # Pared horizontal con R
    'K': ['C', 'W', 'M', 'T'],
    'M': ['D', 'K', 'N', 'F'],
    'N': ['E', 'M'],
    'P': ['G', 'Q'],
    'Q': ['I', 'P', 'R'],
    'R': ['Q', 'T'],     # Pared con W
    'T': ['K', 'R'],     # Pared con F
    'F': ['M']           # Pared con T
}

inicio = 'I'
objetivo = 'F'

# --- 2. ALGORITMO A* (A Estrella) ---

def distancia_manhattan(nodo1, nodo2):
    r1, c1 = coords[nodo1]
    r2, c2 = coords[nodo2]
    return abs(r1 - r2) + abs(c1 - c2)

def obtener_costo(nodo):
    # La casilla W tiene costo 30, el resto 1
    return 30 if nodo == 'W' else 1

def a_star(inicio, objetivo):
    open_set = []
    # heapq ordena por el primer elemento de la tupla (f_score).
    # Si hay empate, compara el segundo elemento (nodo), asegurando el orden alfabético A-Z
    heapq.heappush(open_set, (distancia_manhattan(inicio, objetivo), inicio, 0))
    
    came_from = {}
    g_score = {inicio: 0}
    visitados = []
    
    while open_set:
        f, actual, current_g = heapq.heappop(open_set)
        
        # Evitar re-procesar nodos si llegamos por un camino más costoso
        if actual not in visitados:
            visitados.append(actual)
            
        if actual == objetivo:
            break
            
        for vecino in vecinos[actual]:
            tentative_g = current_g + obtener_costo(vecino)
            
            if vecino not in g_score or tentative_g < g_score[vecino]:
                came_from[vecino] = actual
                g_score[vecino] = tentative_g
                f_score = tentative_g + distancia_manhattan(vecino, objetivo)
                heapq.heappush(open_set, (f_score, vecino, tentative_g))
                
    # Reconstrucción del camino
    camino = []
    curr = objetivo
    while curr in came_from:
        camino.append(curr)
        curr = came_from[curr]
    if camino or inicio == objetivo:
        camino.append(inicio)
    camino.reverse()
    
    return visitados, camino

visitados, camino_optimo = a_star(inicio, objetivo)

# --- 3. INTERFAZ GRÁFICA CON TKINTER ---

TAMANO_CELDA = 80
FILAS = 4
COLUMNAS = 6

COLOR_FONDO = '#FFFFFF'
COLOR_INICIO_FIN = '#FFEB3B' # Amarillo
COLOR_VISITADO = '#BBDEFB'   # Tono tenue para exploración
COLOR_CAMINO = '#4CAF50'     # Tono fuerte para camino
COLOR_PARED = '#FF0000'      # Segmentos rojos

ventana = tk.Tk()
ventana.title("Búsqueda A* sobre Mapa Personalizado")
ventana.resizable(False, False)

# El canvas contendrá toda la estructura visual
canvas = tk.Canvas(ventana, width=COLUMNAS * TAMANO_CELDA, height=FILAS * TAMANO_CELDA, bg='white')
canvas.pack(padx=20, pady=20)

rectangulos = {}

# Dibujar únicamente las celdas válidas del mapa
for letra, (f, c) in coords.items():
    x1 = c * TAMANO_CELDA
    y1 = f * TAMANO_CELDA
    x2 = x1 + TAMANO_CELDA
    y2 = y1 + TAMANO_CELDA
    
    color_actual = COLOR_INICIO_FIN if letra in (inicio, objetivo) else COLOR_FONDO
    
    rect_id = canvas.create_rectangle(x1, y1, x2, y2, fill=color_actual, outline='#757575', width=2)
    rectangulos[letra] = rect_id
    
    canvas.create_text(x1 + TAMANO_CELDA/2, y1 + TAMANO_CELDA/2, text=letra, font=('Arial', 18, 'bold'))

# Dibujar las paredes (Segmentos rojos de la imagen)
paredes_fisicas = [
    ((1, 3), (1, 4)), # Pared Vertical entre C y D
    ((1, 4), (1, 5)), # Pared Vertical entre D y E
    ((2, 2), (3, 2)), # Pared Horizontal entre W y R
    ((3, 3), (3, 4))  # Pared Vertical entre T y F
]

for (f1, c1), (f2, c2) in paredes_fisicas:
    if f1 == f2:  # Vertical
        x = max(c1, c2) * TAMANO_CELDA
        y_inicio = f1 * TAMANO_CELDA
        y_fin = y_inicio + TAMANO_CELDA
        canvas.create_line(x, y_inicio, x, y_fin, fill=COLOR_PARED, width=7)
    else:         # Horizontal
        y = max(f1, f2) * TAMANO_CELDA
        x_inicio = c1 * TAMANO_CELDA
        x_fin = x_inicio + TAMANO_CELDA
        canvas.create_line(x_inicio, y, x_fin, y, fill=COLOR_PARED, width=7)

# --- 4. FUNCIONES DE ANIMACIÓN ---

def animar_busqueda(indice=0):
    if indice < len(visitados):
        nodo_actual = visitados[indice]
        # Evitamos sobrescribir el amarillo brillante de Inicio y Fin durante la búsqueda
        if nodo_actual not in (inicio, objetivo):
            canvas.itemconfig(rectangulos[nodo_actual], fill=COLOR_VISITADO)
            
        ventana.after(400, animar_busqueda, indice + 1)
    else:
        # Una vez termina de buscar, anima la ruta final
        ventana.after(600, animar_camino, 0)

def animar_camino(indice=0):
    if indice < len(camino_optimo):
        nodo_actual = camino_optimo[indice]
        # Aplica el color verde fuerte a toda la ruta encontrada
        canvas.itemconfig(rectangulos[nodo_actual], fill=COLOR_CAMINO)
        ventana.after(300, animar_camino, indice + 1)

# Botón inferior
boton_iniciar = tk.Button(ventana, text="Iniciar Búsqueda A*", font=('Arial', 12, 'bold'), 
                          command=lambda: animar_busqueda(0))
boton_iniciar.pack(pady=10)

ventana.mainloop()
```

6. Desarrolle un agente que emplee una estrategia de búsqueda A* para ir de una casilla a otra evitando la pared representada, pudiendo seleccionar ustedes mismos el inicio y el final. Muestre en una imagen el camino obtenido.

```python
import tkinter as tk
import heapq
from typing import List, Tuple, Dict, Optional, Set, Iterator

# Alias de tipo para representar coordenadas en el plano cartesiano de la cuadrícula: (columna_x, fila_y)
Point = Tuple[int, int]

class Grid:
    """
    Estructura lógica que abstrae la cuadrícula del mapa.
    Mantiene únicamente la información de dimensiones y obstáculos (paredes),
    garantizando que la lógica del espacio esté 100% desacoplada de Tkinter.
    """
    def __init__(self, width: int, height: int, walls: Set[Point]):
        self.width = width        # Total de columnas horizontales de la cuadrícula
        self.height = height      # Total de filas verticales de la cuadrícula
        self.walls = walls        # Conjunto (Set) de coordenadas no transitables para búsquedas $O(1)$

    def in_bounds(self, point: Point) -> bool:
        """
        Verifica si la coordenada evaluada existe físicamente dentro del rango [0, width) y [0, height).
        Evita lecturas fuera de rango o errores en los bordes.
        """
        x, y = point
        return 0 <= x < self.width and 0 <= y < self.height

    def is_passable(self, point: Point) -> bool:
        """
        Comprueba si la casilla es transitable.
        Devuelve False si el punto pertenece al conjunto de paredes registradas.
        """
        return point not in self.walls

    def get_neighbors(self, point: Point) -> List[Point]:
        """
        Genera los nodos adyacentes a la posición actual.
        Aplica la restricción del problema: solo 4 movimientos ortogonales (sin diagonales).
        Filtra automáticamente cualquier casilla fuera de límites o que sea una pared.
        """
        x, y = point
        # Vectores de movimiento: Derecha, Izquierda, Abajo, Arriba
        directions = [(x + 1, y), (x - 1, y), (x, y + 1), (x, y - 1)]
        
        # Filtro de comprensión de lista para descartar posiciones inválidas
        return [p for p in directions if self.in_bounds(p) and self.is_passable(p)]


class AStar:
    """
    Motor puramente algorítmico de búsqueda A*.
    Utiliza funciones estáticas para calcular estimaciones heurísticas y el recorrido.
    """
    @staticmethod
    def heuristic(a: Point, b: Point) -> int:
        """
        Calcula la Distancia Manhattan entre dos puntos.
        Fórmula: h(n) = |x1 - x2| + |y1 - y2|
        
        Es una heurística admisible (nunca sobrestima el costo real) y consistente,
        lo que garantiza que A* siempre encontrará el camino más corto en grillas ortogonales.
        """
        return abs(a[0] - b[0]) + abs(a[1] - b[1])

    @staticmethod
    def search(grid: Grid, start: Point, goal: Point) -> Iterator[Tuple[Optional[Point], Optional[List[Point]], Optional[Dict[Point, int]]]]:
        """
        Ejecuta A* y actúa como un generador (yield) de Python.
        Permite detener temporalmente el cálculo en cada iteración para que la GUI
        pueda animar el nodo recién explorado sin bloquear la ventana principal.
        
        Devuelve una tupla en cada paso:
        - (nodo_explorado_actual, None, None) durante la exploración de nodos.
        - (None, camino_optimo, costos_g) al finalizar exitosamente la búsqueda.
        """
        # Cola de prioridad 'frontier' optimizada con la librería heapq.
        # Almacena elementos en formato (prioridad_f, nodo), ordenando siempre el menor f(n) en la cima.
        frontier = []
        heapq.heappush(frontier, (0, start))
        
        # Diccionario 'came_from' para la reconstrucción posterior del camino (nodo_actual -> nodo_padre)
        came_from: Dict[Point, Optional[Point]] = {start: None}
        
        # Diccionario 'cost_so_far' que registra el costo acumulado exacto g(n) desde el origen hasta 'n'
        cost_so_far: Dict[Point, int] = {start: 0}

        while frontier:
            # Extraemos de la cola de prioridad el nodo con el menor valor estimado f(n)
            _, current = heapq.heappop(frontier)

            # Condición de parada: Si el nodo extraído es el destino, concluimos la exploración
            if current == goal:
                break

            # Evaluamos las casillas vecinas válidas
            for next_node in grid.get_neighbors(current):
                # En este mapa, el costo de avance entre dos casillas contiguas es siempre igual a 1
                new_cost = cost_so_far[current] + 1
                
                # Si el vecino no ha sido visitado O si encontramos una ruta con menor costo g(n)
                if next_node not in cost_so_far or new_cost < cost_so_far[next_node]:
                    cost_so_far[next_node] = new_cost
                    
                    # Cálculo de la ecuación fundamental A*: f(n) = g(n) + h(n)
                    # g(n) = new_cost (costo acumulado real)
                    # h(n) = AStar.heuristic(next_node, goal) (costo estimado hasta el final)
                    priority = new_cost + AStar.heuristic(next_node, goal)
                    
                    # Añadimos a la frontera con su prioridad f(n) calculada
                    heapq.heappush(frontier, (priority, next_node))
                    came_from[next_node] = current
                    
                    # Emitimos el nodo evaluado a la GUI para que sea pintado en la animación
                    if next_node != goal:
                        yield (next_node, None, None)

        # Reconstrucción del camino trazando las referencias hacia atrás (desde Destino hasta Inicio)
        path = []
        if goal in came_from:
            current = goal
            while current != start:
                path.append(current)
                current = came_from[current]
            path.reverse() # Invertimos para obtener la secuencia correcta (Inicio -> Destino)
            
        # Emitimos el resultado final conteniendo el camino óptimo y el registro de costos g(n)
        yield (None, path, cost_so_far)


class PathfindingApp:
    """
    Controlador y renderizador visual mediante Tkinter.
    Administra la captura de eventos de mouse, la actualización del Canvas y la animación.
    """
    def __init__(self, root: tk.Tk, cols: int, rows: int, cell_size: int = 30):
        self.root = root
        self.root.title("A* Pathfinding - Agente de Búsqueda Interactiva")
        
        self.cols = cols
        self.rows = rows
        self.cell_size = cell_size
        
        # Variables de control de estado del ciclo de vida de la aplicación
        self.start_node: Optional[Point] = None
        self.end_node: Optional[Point] = None
        self.is_searching = False
        
        # Carga la disposición por defecto de la pared basada en la imagen de referencia
        self.initial_walls: Set[Point] = self._generate_default_walls()
        self.grid = Grid(self.cols, self.rows, self.initial_walls.copy())
        
        self._setup_ui()
        self.draw_grid()
        self._update_status()

    def _generate_default_walls(self) -> Set[Point]:
        """
        Construye el conjunto de coordenadas que forman el obstáculo en forma de 'L' invertida
        y los bloques superiores en diagonal presentes en la imagen entregada.
        """
        walls = set()
        # Muro vertical principal
        for y in range(4, 17):
            walls.add((10, y))
        # Escalón superior
        walls.add((11, 3))
        walls.add((12, 2))
        # Tramo horizontal en la base
        for x in range(11, 15):
            walls.add((x, 16))
        return walls

    def _setup_ui(self):
        """
        Inicializa los contenedores de la ventana, la barra de botones y el lienzo gráfico.
        """
        control_frame = tk.Frame(self.root, pady=10)
        control_frame.pack(fill=tk.X)
        
        # Etiqueta orientativa para el usuario
        self.status_label = tk.Label(control_frame, text="", font=("Arial", 11, "bold"))
        self.status_label.pack(side=tk.LEFT, padx=10)
        
        # Botones de la barra de control
        self.btn_reset = tk.Button(control_frame, text="Reiniciar", command=self.reset_app, bg="#f44336", fg="white", font=("Arial", 9, "bold"))
        self.btn_reset.pack(side=tk.RIGHT, padx=5)
        
        self.btn_search = tk.Button(control_frame, text="Buscar camino", command=self.start_search, state=tk.DISABLED, bg="#4CAF50", fg="white", font=("Arial", 9, "bold"))
        self.btn_search.pack(side=tk.RIGHT, padx=5)
        
        # Lienzo (Canvas) para el dibujo interactivo de la grilla
        canvas_width = self.cols * self.cell_size
        canvas_height = self.rows * self.cell_size
        self.canvas = tk.Canvas(self.root, width=canvas_width, height=canvas_height, bg="white")
        self.canvas.pack(padx=10, pady=10)
        
        # Captura de clics del mouse sobre el lienzo
        self.canvas.bind("<Button-1>", self.handle_click)

    def draw_grid(self):
        """
        Redibuja el mapa completo limpiando objetos previos en el canvas.
        Dibuja los bordes de la retícula y aplica colores según el tipo de celda.
        """
        self.canvas.delete("all")
        
        for x in range(self.cols):
            for y in range(self.rows):
                # Conversión de índices de columna/fila a coordenadas de píxeles (x0, y0, x1, y1)
                x0 = x * self.cell_size
                y0 = y * self.cell_size
                x1 = x0 + self.cell_size
                y1 = y0 + self.cell_size
                
                point = (x, y)
                color = "white"
                text = ""
                text_color = "black"
                
                # Discriminación de tipos de casillas para aplicar color
                if point in self.grid.walls:
                    color = "black"
                elif point == self.start_node:
                    color = "#4169E1" # Azul Real
                    text = "I"
                    text_color = "white"
                elif point == self.end_node:
                    color = "#FFD700" # Amarillo brillante
                    text = "F"
                    text_color = "black"
                    
                # Dibujado de la casilla física
                self.canvas.create_rectangle(x0, y0, x1, y1, fill=color, outline="#D3D3D3", tags=f"cell_{x}_{y}")
                
                # Renderizado del identificador textual (Inicio 'I' o Fin 'F')
                if text:
                    self.canvas.create_text(
                        x0 + self.cell_size / 2, 
                        y0 + self.cell_size / 2, 
                        text=text, 
                        fill=text_color, 
                        font=("Arial", 10, "bold"),
                        tags=f"text_{x}_{y}"
                    )

    def color_cell(self, point: Point, color: str):
        """
        Permite modificar el color de fondo de un rectángulo específico mediante su tag único.
        Evita tener que redibujar toda la grilla en cada frame de la animación.
        """
        x, y = point
        tag = f"cell_{x}_{y}"
        self.canvas.itemconfig(tag, fill=color)

    def draw_cost_text(self, point: Point, cost: int, text_color: str = "#8B0000"):
        """
        NUEVA FUNCIONALIDAD: Dibuja sobre la casilla el valor numérico del costo acumulado g(n)
        definitivo que le tomó al algoritmo llegar a ese punto en la ruta óptima.
        
        Usa por defecto el color rojo oscuro (#8B0000) para destacar claramente sobre el verde.
        """
        x, y = point
        x_center = (x * self.cell_size) + (self.cell_size / 2)
        y_center = (y * self.cell_size) + (self.cell_size / 2)
        
        # Elimina cualquier texto previo presente en esa celda para evitar solapamientos
        self.canvas.delete(f"text_{x}_{y}")
        
        # Inserta el costo acumulado como un elemento textual centrado
        self.canvas.create_text(
            x_center, 
            y_center, 
            text=str(cost), 
            fill=text_color, 
            font=("Arial", 9, "bold"),
            tags=f"text_{x}_{y}"
        )

    def handle_click(self, event):
        """
        Manejador de clics del usuario:
        1er Clic -> Establece Inicio ('I')
        2do Clic -> Establece Destino ('F')
        """
        if self.is_searching:
            return # Bloquea cambios durante la ejecución del algoritmo
            
        # Transforma píxeles a coordenadas de la cuadrícula
        x = event.x // self.cell_size
        y = event.y // self.cell_size
        point = (x, y)
        
        # Ignora clics fuera de rango o dirigidos hacia paredes
        if not self.grid.in_bounds(point) or not self.grid.is_passable(point):
            return
            
        # Asignación del punto inicial
        if self.start_node is None:
            self.start_node = point
        # Asignación del punto final (solo si no coincide con el inicio)
        elif self.end_node is None and point != self.start_node:
            self.end_node = point
            self.btn_search.config(state=tk.NORMAL) # Habilita el botón de búsqueda
            
        self.draw_grid()
        self._update_status()

    def _update_status(self):
        """
        Actualiza el texto descriptivo del estado superior para orientar al usuario.
        """
        if self.start_node is None:
            self.status_label.config(text="1. Clic para seleccionar la casilla de inicio (I)", fg="#1E90FF")
        elif self.end_node is None:
            self.status_label.config(text="2. Clic para seleccionar la casilla de destino (F)", fg="#B8860B")
        else:
            self.status_label.config(text="3. Presiona 'Buscar camino' para ejecutar A*", fg="#2E8B57")

    def start_search(self):
        """
        Prepara el estado visual e inicia el consumo del generador A*.
        """
        if not self.start_node or not self.end_node:
            return
            
        self.is_searching = True
        self.btn_search.config(state=tk.DISABLED)
        self.status_label.config(text="Buscando camino con A*...", fg="#708090")
        
        # Limpia ejecuciones anteriores manteniendo los obstáculos e hitos
        self.draw_grid()
        
        # Instancia el generador iterativo
        self.search_generator = AStar.search(self.grid, self.start_node, self.end_node)
        self.animate_search()

    def animate_search(self):
        """
        Realiza llamadas asíncronas no bloqueantes con `root.after` consumiendo el generador de A*.
        Pinta los nodos explorados en la animación y, al finalizar, muestra la ruta y los costos finales.
        """
        try:
            explored_node, path, cost_so_far = next(self.search_generator)
            
            if explored_node:
                # Transición visual: Pinta de azul claro las casillas exploradas durante la búsqueda
                self.color_cell(explored_node, "#E0F7FA")
                self.root.after(15, self.animate_search) # Re-agenda el siguiente paso a los 15 milisegundos
                
            elif path is not None and cost_so_far is not None:
                # Fase final: Búsqueda completada
                self.is_searching = False
                if path:
                    # Dibuja cada casilla del camino final encontrado
                    for node in path:
                        if node == self.end_node:
                            # En la casilla final 'F', pintamos el texto del costo acumulado total en color negro
                            self.draw_cost_text(node, cost_so_far[node], text_color="#000000")
                        else:
                            # Pinta la casilla en verde brillante para destacar la ruta óptima
                            self.color_cell(node, "#32CD32")
                            # Muestra el costo $g(n)$ final con el cual se quedó esa casilla en color rojo oscuro destacado
                            self.draw_cost_text(node, cost_so_far[node], text_color="#8B0000")
                            
                    # Muestra el costo 0 en la casilla de inicio ('I') en color blanco
                    if self.start_node:
                        self.draw_cost_text(self.start_node, 0, text_color="#FFFFFF")

                    total_cost = cost_so_far[self.end_node]
                    self.status_label.config(
                        text=f"¡Camino óptimo encontrado! Costo total = {total_cost} pasos", 
                        fg="green"
                    )
                else:
                    self.status_label.config(text="No existe un camino posible", fg="red")
                    
        except StopIteration:
            self.is_searching = False

    def reset_app(self):
        """
        Restaura la aplicación a su estado inicial para permitir realizar nuevas pruebas.
        """
        self.start_node = None
        self.end_node = None
        self.is_searching = False
        self.btn_search.config(state=tk.DISABLED)
        self.grid.walls = self.initial_walls.copy()
        self.draw_grid()
        self._update_status()

if __name__ == "__main__":
    # Inicialización de Tkinter y ejecución de la aplicación
    root = tk.Tk()
    app = PathfindingApp(root, cols=22, rows=20, cell_size=25)
    root.mainloop()
```
