## Ejercicios Teóricos

1. ¿Qué diferencia hay entre una estrategia de búsqueda Informada y una estrategia de búsqueda No Informada?

Una estrategia de búsqueda no informada no tiene información adicional sobre qué tan cerca o en que dirección está un estado de la solución, esta explora de manera sistemática, puede tardar más tiempo, avanza evaluando todas las opciones disponibles por igual. En cambio la estrategia de búsqueda informada utiliza información adicional, llamada heurística, para estimar que tan conveniente es seguir por un camino determinado, permite priorizar los caminos más convenientes y llegar al resultado de manera más rápida.

2. ¿Qué es una heurística y para qué sirve?

Una heurística es una función matemática h(n) que dado cierto nodo en el espacio, devuelve una estimación del costo mínimo restante para alcanzar un estado meta desde n. Sirve para guiar la búsqueda, reducir costo computacional, mejorar eficiencia, balancear precisión y rapidez, etc.

3. ¿Es posible que un algoritmo de búsqueda no tenga solución?

Si, si es posible que un algoritmo de búsqueda no tenga solución, esto puede pasar en los algoritmos infinitos, cuando no existe solución o un camino válido para llegar a la solución, que se haya definido incorrectamente el problema, que el algoritmo tenga limitaciones o que los recursos sean insuficientes.
